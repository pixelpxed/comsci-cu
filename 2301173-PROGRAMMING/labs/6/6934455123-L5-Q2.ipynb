## Q2
จงเขียนโปรแกรมที่อ่านเฉลยของข้อสอบปรนัยจากไฟล์ `sol<n>.txt` (ข้อสอบมี 10 ข้อ)
และอ่านคำตอบของนักเรียนจากไฟล์ `exam<n>.txt` ที่มีคำตอบของแต่ละคนอยู่ใน 1 บรรทัด
(ถือว่านักเรียนตอบครบทุกข้อ และมีนักเรียน 8 คน)

ให้เขียนฟังก์ชัน ทุกฟังก์ชันรับชื่อไฟล์ 2 ค่า คือ `solution_filename` กับ `exam_filename` และคืนค่าเป็นลิสต์

- `student_scores(solution_filename, exam_filename)` คะแนนของนักเรียนแต่ละคน
- `question_averages(solution_filename, exam_filename)` สัดส่วนของนักเรียนที่ตอบถูกในแต่ละข้อ
- `hardest_questions(solution_filename, exam_filename)` หมายเลขข้อที่สัดส่วนตอบถูกน้อยสุด
- `easiest_questions(solution_filename, exam_filename)` หมายเลขข้อที่สัดส่วนตอบถูกมากสุด

หมายเลขข้อเริ่มจาก 1 ถ้ามีหลายข้อที่ค่าเท่ากัน ให้คืนทุกข้อเรียงจากน้อยไปมาก


```python
def read_file(file):
    file_raw = open(file)
    file_contents = file_raw.readlines()
    file_raw.close()

    for i in range(len(file_contents)):
        file_contents[i] = file_contents[i].strip().split(" ")

    return file_contents

def student_scores(solution_filename, exam_filename):
    solutions = read_file(solution_filename)[0]
    answers = read_file(exam_filename)
    scores = []

    for answer in answers:
        score = 0
        for i in range(len(solutions)):
            if (answer[i] == solutions[i]):
                score = score + 1
        scores.append(score)

    return scores

def question_averages(solution_filename, exam_filename):
    solutions = read_file(solution_filename)[0]
    answers = read_file(exam_filename)

    answers_count = len(answers)

    averages = []

    for question in range(len(solutions)):
        sum_correct_answers = 0
        
        for student in range(answers_count):
            if (answers[student][question] == solutions[question]):
                sum_correct_answers = sum_correct_answers + 1

        averages.append(sum_correct_answers / answers_count)

    return averages

def hardest_questions(solution_filename, exam_filename):
    data = question_averages(solution_filename, exam_filename)
    hardest_avg = min(data)
    hardest_list = []

    for i in range(len(data)):
        if (data[i] == hardest_avg):
            hardest_list.append(i + 1)

    return hardest_list



def easiest_questions(solution_filename, exam_filename):
    data = question_averages(solution_filename, exam_filename)
    easiest_avg = max(data)
    easiest_list = []

    for i in range(len(data)):
        if (data[i] == easiest_avg):
            easiest_list.append(i + 1)

    return easiest_list
```

```python
"""ตรวจสอบคะแนนของนักเรียน"""
assert student_scores('sol1.txt', 'exam1.txt') == [9, 6, 9, 8, 7, 7, 9, 8]
assert student_scores('sol1.txt', 'exam2.txt') == [8, 9, 5, 7, 10, 9, 6, 9]
assert student_scores('sol2.txt', 'exam1.txt') == [6, 5, 6, 8, 4, 7, 6, 5]
assert student_scores('sol2.txt', 'exam2.txt') == [5, 6, 4, 4, 7, 6, 9, 6]
print('ok')
```

```python
"""ตรวจสอบคะแนนเฉลี่ยของข้อสอบแต่ละข้อ"""
assert question_averages('sol1.txt', 'exam1.txt') == [0.625, 1.0, 0.5, 0.875, 0.875, 1.0, 0.75, 0.75, 0.75, 0.75]
assert question_averages('sol1.txt', 'exam2.txt') == [0.75, 1.0, 0.5, 0.875, 0.875, 0.75, 0.75, 0.875, 0.75, 0.75]
assert question_averages('sol2.txt', 'exam1.txt') == [0.125, 1.0, 0.5, 0.875, 0.875, 0.0, 0.75, 0.25, 0.75, 0.75]
assert question_averages('sol2.txt', 'exam2.txt') == [0.125, 1.0, 0.5, 0.875, 0.875, 0.125, 0.75, 0.125, 0.75, 0.75]
print('ok')
```

```python
"""ตรวจสอบข้อสอบที่ยากที่สุด"""
assert hardest_questions('sol1.txt', 'exam1.txt') == [3]
assert hardest_questions('sol1.txt', 'exam2.txt') == [3]
assert hardest_questions('sol2.txt', 'exam1.txt') == [6]
assert hardest_questions('sol2.txt', 'exam2.txt') == [1, 6, 8]
print('ok')
```

```python
"""ตรวจสอบข้อสอบที่ง่ายที่สุด"""
assert easiest_questions('sol1.txt', 'exam1.txt') == [2, 6]
assert easiest_questions('sol1.txt', 'exam2.txt') == [2]
assert easiest_questions('sol2.txt', 'exam1.txt') == [2]
assert easiest_questions('sol2.txt', 'exam2.txt') == [2]
print('ok')
```
