## Q3
จงเขียนฟังก์ชันสำหรับจัดการลิสต์สินค้า ฟังก์ชันที่แก้ลิสต์ต้องแก้ `item_list` เดิม แล้วคืนค่าข้อความตามที่กำหนด

**เพิ่มสินค้าท้ายลิสต์**
`add_item(item_list, item)`
การทำงาน
- ถ้ายังไม่มีสินค้านั้นในลิสต์ ให้เพิ่มไว้ท้ายลิสต์ แล้วคืนค่า `"Item has been added"`
- ถ้ามีอยู่แล้ว ไม่ต้องเพิ่มซ้ำ คืนค่า `"Item is already in the list"`

**เปลี่ยนชื่อสินค้า**
`change_item(item_list, old_item, new_item)`
การทำงาน
- ถ้ามี `old_item` ในลิสต์ ให้เปลี่ยนเป็น `new_item` แล้วคืนค่า `"Item has been changed"`
- ถ้าไม่มี คืนค่า `"Item is not in the list"`

**แทรกสินค้าตามตำแหน่ง**
`insert_item(item_list, item, location)`
การทำงาน
- แทรก `item` ที่ตำแหน่ง `location` ตาม `list.insert` ไม่ต้องตรวจว่าซ้ำหรือไม่
- คืนค่า `"Item has been inserted"`

**ลบสินค้า**
`remove_item(item_list, item)`
การทำงาน
- ถ้ามีในลิสต์ ให้ลบตัวแรกที่เจอ แล้วคืนค่า `"Item has been removed"`
- ถ้าไม่มี คืนค่า `"This item is not in the list"`

**แสดงสินค้า**
`show_items(item_list)`
การทำงาน
- ถ้าลิสต์ว่าง คืนค่า `"The list is currently empty"`
- ถ้ามีสินค้า คืนค่าลิสต์นั้น

ดูการทดสอบของฟังก์ชันเหล่านี้ได้ในเซลล์ด้านล่าง


### ฟังก์ชัน Main()

ให้เขียนฟังก์ชัน `Main()` เพื่อแสดงเมนู รับตัวเลือกจากผู้ใช้ แล้วเรียกฟังก์ชันด้านบนตามเมนูที่เลือก

พิมพ์เมนูตอนเริ่มครั้งเดียว จากนั้นวนรับ `Select menu:` จนกว่าจะเลือก 6

1. Show items พิมพ์ค่าที่ได้จาก `show_items`
2. Add item รับ `Enter item:` แล้วเรียก `add_item`
3. Change item รับ `Enter old item:` แล้ว `Enter new item:` จากนั้นเรียก `change_item`
4. Insert item รับ `Enter item:` แล้ว `Enter location index:` (จำนวนเต็ม ตำแหน่งเริ่มที่ 0 ตาม `list.insert`) จากนั้นเรียก `insert_item`
5. Remove item รับ `Enter item:` แล้วเรียก `remove_item`
6. Exit พิมพ์ `Goodbye!` แล้วจบ

เมนู 1 ถึง 5 ให้พิมพ์ค่าที่ฟังก์ชันคืนมา หลังจบแต่ละคำสั่งยกเว้น Exit ให้เว้นบรรทัดว่างหนึ่งบรรทัดตามตัวอย่าง


ตัวอย่างด้านล่างคือหน้าจอตอนรันจริง


### ตัวอย่างการทำงานของ Main()

````python
def add_item(item_list, item):
    if (item in item_list):
        return "Item is already in the list"
    item_list.append(item)
    return "Item has been added"

def change_item(item_list, old_item, new_item):
    for i in range(len(item_list)):
        if (item_list[i] == old_item):
            item_list[i] = new_item
            return "Item has been changed"
    return "Item is not in the list"

def insert_item(item_list, item, location):
    item_list.insert(location, item)
    return "Item has been inserted"

def remove_item(item_list, item):
    if (item in item_list):
        item_list.remove(item)
        return "Item has been removed"
    return "This item is not in the list"

def show_items(item_list):
    if (len(item_list) != 0):
        return item_list
    return "The list is currently empty"

def Main():
    action = ""
    list = []

    print("""=== Shopping List Menu ===
1. Show items
2. Add item
3. Change item
4. Insert item
5. Remove item
6. Exit""")
    
    while (action != "6"):
        action = input("Choose Your Action: ")

        if (action == "6"):
            return print("Goodbye!")

        if (action == "1"):
            print(show_items(list))
        elif (action == "2"):
            item_add = input("Enter item: ")
            print(add_item(list, item_add))
        elif (action == "3"):
            old_item = input("Enter old item: ")
            new_item = input("Enter new item: ")
            print(change_item(list, old_item, new_item))
        elif (action == "4"):
            item_insert = input("Enter item: ")
            location = int(input("Enter location: "))
            print(insert_item(list, item_insert, location))
        elif (action == "5"):
            item_remove = input("Enter item: ")
            print(remove_item(list, item_remove))
        print()
````

````python
"""ตรวจสอบฟังก์ชัน show_items"""
items = []
assert show_items(items) == "The list is currently empty"
items = ['Pen', 'Eraser']
assert show_items(items) == ['Pen', 'Eraser']
print('ok')
````

````python
"""ตรวจสอบฟังก์ชัน add_item"""
items = []
assert add_item(items, 'Pen') == "Item has been added"
assert add_item(items, 'Eraser') == "Item has been added"
assert add_item(items, 'Pen') == "Item is already in the list"
assert items == ['Pen', 'Eraser']
print('ok')
````

````python
"""ตรวจสอบฟังก์ชัน change_item"""
items = ['Pen', 'Eraser']
assert change_item(items, 'Rubber', 'Bag') == "Item is not in the list"
assert items == ['Pen', 'Eraser']
assert change_item(items, 'Eraser', 'Ruler') == "Item has been changed"
assert items == ['Pen', 'Ruler']
print('ok')
````

````python
"""ตรวจสอบฟังก์ชัน insert_item"""
items = ['Pen', 'Eraser']
assert insert_item(items, 'Pencil', 0) == "Item has been inserted"
assert items == ['Pencil', 'Pen', 'Eraser']
print('ok')
````

````python
"""ตรวจสอบฟังก์ชัน remove_item"""
items = ['Pen', 'Eraser', 'Pencil']
assert remove_item(items, 'Sharpener') == "This item is not in the list"
assert items == ['Pen', 'Eraser', 'Pencil']
assert remove_item(items, 'Pen') == "Item has been removed"
assert items == ['Eraser', 'Pencil']
print('ok')
````
