## Q1
จงเขียนฟังก์ชันอ่านเวกเตอร์ 2 เวกเตอร์จากไฟล์ แล้วหา dot product

- `is_equal(v1, v2)` ตรวจว่าเวกเตอร์ยาวเท่ากันหรือไม่ คืนค่า `True` หรือ `False`
- `dot(v1, v2)` หา dot product ของลิสต์ตัวเลขที่ขนาดเท่ากัน คืนค่าผลรวม
- `convert_to_float(v)` แปลงสมาชิกในลิสต์จาก string เป็น `float` แล้วคืนลิสต์นั้น
- `read_file_vector(filename)` อ่าน 2 บรรทัดแรกของไฟล์ คืนค่าเป็น tuple `(v1, v2)` ที่ยังเป็นลิสต์ของ string
- `vector_dot(filename)` อ่านไฟล์แล้วแปลงเป็น float ถ้าขนาดเท่ากันคืนค่า dot product เป็น `float` ถ้าไม่เท่ากันคืนค่า `"Incompatible size"`


```python
def is_equal(v1, v2):
    return len(v1) == len(v2)


def dot(v1, v2):    
    sum = 0
    for i in range(len(v1)):
        sum = sum + (v1[i] * v2[i])
    return sum

def convert_to_float(v):
    result = []
    for string in v:
        result.append(float(string))
    return result

def read_file_vector(filename):
    file_raw = open(filename)
    file_contents = file_raw.readlines()
    file_raw.close()

    result = []
    for line in file_contents:
        result.append(line.strip().split(" "))

    return (result[0], result[1])

def vector_dot(filename):
    vectors = read_file_vector(filename)

    # Clean, converts data to preferred type.
    v1, v2 = convert_to_float(vectors[0]), convert_to_float(vectors[1])

    # Stop function if vector is invalid anyways.
    if (is_equal(v1, v2) == False):
        return "Incompatible size"
    return dot(v1, v2)
```

```python
"""ตรวจสอบฟังก์ชัน is_equal"""
assert is_equal([6.0, 4.0, 9.0, 6.0, -4.0, 7.0], [3.0, 7.0, -3.0, 9.0, -5.0, 1.0]) == True
assert is_equal([1.0, 2.0, 3.0], [4.0, -5.1, -99.2]) == True
assert is_equal([1.0, 2.0], [4.0, -5.1, -99.2]) == False
print('ok')
```

```python
"""ตรวจสอบฟังก์ชัน dot"""
assert dot([6.0, 4.0, 9.0, 6.0, -4.0, 7.0], [3.0, 7.0, -3.0, 9.0, -5.0, 1.0]) == 100.0
assert dot([1.0, 2.0, 3.0], [4.0, -5.1, -99.2]) == -303.8
print('ok')
```

```python
"""ตรวจสอบฟังก์ชัน convert_to_float"""
assert convert_to_float(['1', '2', '3']) == [1.0, 2.0, 3.0]
assert convert_to_float(['-55', '-81', '0.01']) == [-55.0, -81.0, 0.01]
print('ok')
```

```python
"""ตรวจสอบฟังก์ชัน read_file_vector"""
assert read_file_vector('vector2.txt') == (['6', '4', '9', '6', '-4', '7'], ['3', '7', '-3'])
assert read_file_vector('vector3.txt') == (['5.5', '4.5', '9.5', '6', '-4', '7'], ['3', '7.1', '-3.5', '5', '6', '9'])
print('ok')
```

```python
"""ตรวจสอบฟังก์ชัน vector_dot เมื่อเวกเตอร์มีขนาดเท่ากัน"""
assert vector_dot('vector1.txt') == 100.0
assert vector_dot('vector3.txt') == 84.2
print('ok')
```

```python
"""ตรวจสอบฟังก์ชัน vector_dot เมื่อเวกเตอร์มีขนาดไม่เท่ากัน"""
assert vector_dot('vector2.txt') == "Incompatible size"
assert vector_dot('vector4.txt') == "Incompatible size"
print('ok')
```
