## Q3
จงเขียนโปรแกรมเพื่อตัดสินผลการประกวดร้องเพลง โดยให้โปรแกรมอ่านคะแนนที่ผู้เข้าประกวดได้รับจากคณะกรรมการ <br>
(เต็ม 10 คะแนนสำหรับกรรมการแต่ละท่าน) และผู้ชมในห้องส่ง (เต็ม 70 คะแนน) <br>
จากไฟล์ score\<n\>.txt (แต่ละบรรทัดประกอบด้วยรหัสผู้ประกวดร้องเพลง คะแนนจากกรรมการคนที่ 1 คะแนนจากกรรมการคนที่ 2 คะแนนจากกรรมการคนที่ 3 และคะแนนจากผู้ชมในห้องส่ง) <br>

กำหนดให้เขียนฟังก์ชัน `get_dict(file_name)` ที่มี parameter เป็นชื่อ file แล้วสร้างและคืนค่า dictionary จากการอ่านไฟล์ โดยมี
- key เป็นสตริงของรหัสผู้เข้าประกวด 
- value เป็นผลรวมคะแนนทั้งหมด

จงเขียนฟังก์ชัน `find_top3(scores)` ที่มี parameter เป็น dictionary ของผู้เข้าประกวดและคะแนน แล้วคืนค่าเป็น list ของผู้มีคะแนนสูงสุดการประกวด 3 อันดับแรก กำหนดให้ไม่มีผู้ใดได้คะแนนเท่ากัน


### ตัวอย่างไฟล์  score1.txt: 

CU01 8 8 9 65 <br>
CU02 6 6 6 45 <br>
CU03 8 8 7 55 <br>
CU04 6 8 6 57 <br>
CU05 9 9 9 66 <br>


### ตัวอย่างไฟล์ score2.txt: 

CU01 8 8 9 65 <br>
CU02 6 6 6 45 <br>
CU03 8 7 7 55 <br>
CU04 6 8 6 57 <br>
CU05 9 9 9 66 <br>
CU06 8 8 8 62 <br>
CU07 10 8 9 67 <br>


### ตัวอย่างไฟล์ score3.txt: 

CU01 8 8 9 65 <br>
CU02 6 6 6 45 <br>
CU03 8 7 7 55 <br>
CU04 6 8 6 57 <br>
CU05 9 9 9 60 <br>
CU06 8 8 8 62 <br>
CU07 10 8 9 67 <br>
CU08 8 7 9 60 <br>
CU09 7 7 7 58 <br>
CU10 9 8 9 60 <br>


### ตัวอย่างการทำงาน: 

find_top3(get_dict(**'score1.txt'**))<br>
['CU05', 'CU01', 'CU03']


### ตัวอย่างการทำงาน: 

find_top3(get_dict(**'score2.txt'**))<br>
['CU07', 'CU05', 'CU01']


### ตัวอย่างการทำงาน: 

find_top3(get_dict(**'score3.txt'**))<br>
['CU07', 'CU01', 'CU05']


```python
def get_dict(file_name):
    file_raw = open(file_name, "r")
    file_contents = file_raw.readlines()

    scores = {}
    for line in file_contents:
        line = line.split()
        scores[line[0]] = sum([
            int(line[1]),
            int(line[2]),
            int(line[3]),
            int(line[4])
        ])

    file_raw.close()
    return scores


def find_top3(scores):
    sorted_scores = sorted(scores.items(), key = lambda x: x[1], reverse = True)
    return [
        sorted_scores[0][0],
        sorted_scores[1][0],
        sorted_scores[2][0]
    ]
```

```python
'''ทดสอบฟังก์ชัน get_dict(file_name)'''
assert get_dict('score1.txt') == {'CU01': 90, 'CU02': 63, 'CU03': 78, 'CU04': 77, 'CU05': 93}
assert get_dict('score2.txt') == {
    'CU01': 90,
    'CU02': 63,
    'CU03': 77,
    'CU04': 77,
    'CU05': 93,
    'CU06': 86,
    'CU07': 94,
}
assert get_dict('score3.txt') == {
    'CU01': 90,
    'CU02': 63,
    'CU03': 77,
    'CU04': 77,
    'CU05': 87,
    'CU06': 86,
    'CU07': 94,
    'CU08': 84,
    'CU09': 79,
    'CU10': 86,
}
print('ok')
```

```python
'''ทดสอบฟังก์ชัน find_top3()'''
assert find_top3({'A': 80, 'B': 90, 'C': 70, 'D': 50, 'E': 95, 'F': 86}) == ['E', 'B', 'F']
assert find_top3({'A': 97, 'B': 68, 'C': 96, 'D': 88, 'E': 85, 'F': 86}) == ['A', 'C', 'D']
print('ok')
```

```python
'''ทดสอบ find_top3 จากไฟล์'''
assert find_top3(get_dict('score1.txt')) == ['CU05', 'CU01', 'CU03']
assert find_top3(get_dict('score2.txt')) == ['CU07', 'CU05', 'CU01']
assert find_top3(get_dict('score3.txt')) == ['CU07', 'CU01', 'CU05']
print('ok')
```
