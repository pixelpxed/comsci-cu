#### Q4
จงเขียนโปรแกรมที่อ่านไฟล์ sports1.txt ที่เก็บข้อมูลชื่อกีฬาและจำนวนพนักงานที่ไปเล่นกีฬาในหนึ่งสัปดาห์มาเก็บใน dictionary <br>
แล้วหาว่ากีฬาใดที่มีพนักงานเล่นไม่ต่ำกว่า 15 คน และจำนวนพนักงานที่มาเล่นกีฬาทั้งหมด <br>

กำหนดให้เขียนฟังก์ชัน `get_dict(file_name)` ที่มี parameter เป็นชื่อ file แล้วสร้างและคืนค่า dictionary จากการอ่านไฟล์ โดยมี
- key เป็นสตริงของชื่อกีฬา 
- value เป็นผลรวมของจำนวนพนักงานที่เล่นกีฬาประเภทนั้น ๆ

และเขียนฟังก์ชัน `popular_sports(sports)` ที่รับ dictionary แล้วคืนค่าเป็น list ของชื่อกีฬาที่มีพนักงานเล่นอย่างน้อย 15 คน <br>
และเขียนฟังก์ชัน `total_players(sports)` ที่รับ dictionary แล้วคืนค่าจำนวนพนักงานที่มาเล่นกีฬาทั้งหมดในหนึ่งสัปดาห์ ดังแสดงในตัวอย่างการทำงาน


### ตัวอย่างไฟล์ sports1.txt:
swimming 5<br>
yoga 10<br>
badminton 8<br>
yoga 8<br>
badminton 10


### ตัวอย่างไฟล์ sports2.txt:
swimming 5<br>
yoga 10<br>
badminton 8<br>
yoga 8<br>
football 19


### ตัวอย่างการทำงาน: 
popular_sports(get_dict(**'sports1.txt'**))<br>
yoga <br>
badminton <br>
total_players(get_dict(**'sports1.txt'**)) <br>
41


## ตัวอย่างการทำงาน: 
popular_sports(get_dict(**'sports2.txt'**))<br>
yoga <br>
football <br>
total_players(get_dict(**'sports2.txt'**)) <br>
50


```python
def get_dict(file_name):
    file_raw = open(file_name, "r")
    file_contents = file_raw.readlines()

    sport_stats = {}

    for line in file_contents:
        line = line.split()

        if (line[0] in sport_stats):
            sport_stats[line[0]] += int(line[1])
        else:
            sport_stats[line[0]] = int(line[1])

    file_raw.close()
    return sport_stats


def popular_sports(sports):
    list_popular = []
    list_sports = list(sports.keys())

    for sport in list_sports:
        if (sports[sport] >= 15):
            list_popular.append(sport)

    return list_popular


def total_players(sports):
    sum = 0
    list_sports = list(sports.keys())

    for sport in list_sports:
        sum += sports[sport]

    return sum
```

```python
'''ตรวจสอบ popular_sports และ total_players'''
assert popular_sports(get_dict('sports1.txt')) == ['yoga', 'badminton']
assert total_players(get_dict('sports1.txt')) == 41
assert popular_sports(get_dict('sports2.txt')) == ['yoga', 'football']
assert total_players(get_dict('sports2.txt')) == 50
print('ok')
```
