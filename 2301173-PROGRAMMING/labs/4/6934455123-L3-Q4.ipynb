- คะแนนเฉลี่ยของนักเรียนทั้งหมด
- คะแนนเฉลี่ยของนักเรียนที่สอบผ่าน (ได้คะแนนอย่างน้อย 5)
- คะแนนเฉลี่ยของนักเรียนที่สอบไม่ผ่าน (ได้คะแนนต่ำกว่า 5)
- คะแนนสูงสุดของนักเรียนทั้งหมด

### ตัวอย่างการทำงาน: 

Number of students : **3** <br>
Student 1 : **1** <br>
Student 2 : **2.5** <br>
Student 3 : **3.5** <br>
Average score : 2.3333333333333335 <br>
Average passing score : No student passed <br>
Average failing score : 2.3333333333333335 <br>
Highest score : 3.5 <br>

```python
def run():
    # YOUR CODE HERE
    num_students = int(input("Number of students: "))
    while (num_students < 1):
        num_students = int(input("Incorrect input. Number of students: "))

    high_score = float(0)
    pass_score = [0, 0] # [score_sum, count]
    fail_score = [0, 0] # [score_sum, count]
    sum = float(0)

    for i in range(num_students):
        score = float(input(f"Student {i + 1}: "))

        if (score > high_score): high_score = score

        if (score >= 5): pass_score = [pass_score[0] + score, pass_score[1] + 1]
        else:            fail_score = [fail_score[0] + score, fail_score[1] + 1]

        sum = sum + score

    if (fail_score[1] == 0): fail_avg = "No student failed"
    else:                    fail_avg = fail_score[0] / fail_score[1]

    if (pass_score[1] == 0): pass_avg = "No student passed"
    else:                    pass_avg = pass_score[0] / pass_score[1]

    print(f"""Average score : {sum / num_students}
Average passing score : {pass_avg}
Average failing score : {fail_avg}
Highest score : {high_score}""")
```

```python
'''ตรวจสอบผลลัพธ์ของโปรแกรมด้วยการรันที่ cell นี้'''

run()
```

```python
'''ตรวจสอบผลการคำนวณ'''

from unittest.mock import patch
from io import StringIO


def test(inputs, expected_output):
    with patch('builtins.input', side_effect=inputs), patch('sys.stdout', new_callable=StringIO) as mock_stdout:
        run()
        assert mock_stdout.getvalue() == expected_output, mock_stdout.getvalue()
```
