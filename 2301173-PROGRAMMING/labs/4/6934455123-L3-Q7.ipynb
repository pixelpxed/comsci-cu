### ตัวอย่างการทำงาน:

Day 3 : **35.0** <br>

Day 5 : **34.0** <br>

Day 6 : **35.6** <br>
Day 7 : **34** <br>
Temperature dropped on day 7

```python
def run():
    history = []

    for i in range(7):
        weather_temp = float(input(f"Day {i + 1}: "))
        history.append(weather_temp)

        if (len(history) > 1 and weather_temp < history[i - 1]):
            print(f"Temperature dropped on day {i + 1}")

    pass
```

```python
'''ตรวจสอบผลลัพธ์ของโปรแกรมด้วยการรันที่ cell นี้'''

run()
```

```python
'''ตรวจสอบผลการคำนวณ'''

from unittest.mock import patch
from io import StringIO


def test(inputs, expected_output):
    with patch('builtins.input', side_effect=inputs), patch('sys.stdout', new_callable=StringIO) as mock_stdout:
        run()
        assert mock_stdout.getvalue() == expected_output, mock_stdout.getvalue()
```

```python
'''ตรวจสอบผลการคำนวณและการแสดงผลลัพธ์ให้เหมือนกับโจทย์ที่กำหนดให้'''

'''Drop 1 days'''
test(['32.6', '34.4', '35.0', '34.6', '35', '35.6', '36'], 'Temperature dropped on day 4\n')
test(['32.6', '31.4', '35.0', '35.6', '35.8', '35.9', '36'], 'Temperature dropped on day 2\n')
test(['32.6', '33.4', '35.0', '35.6', '35.8', '35.9', '34'], 'Temperature dropped on day 7\n')
print('ok')
```

```python
'''ตรวจสอบผลการคำนวณและการแสดงผลลัพธ์ให้เหมือนกับโจทย์ที่กำหนดให้'''

'''Drop 2 days'''
test(['32.6', '34.4', '35.0', '34.6', '35.2', '35.6', '34'], 'Temperature dropped on day 4\nTemperature dropped on day 7\n')
test(['33.6', '34.4', '32.0', '35.6', '37.2', '31.6', '35'], 'Temperature dropped on day 3\nTemperature dropped on day 6\n')
test(['43.6', '40.4', '42.0', '45.6', '30.2', '31.6', '35'], 'Temperature dropped on day 2\nTemperature dropped on day 5\n')
print('ok')
```

```python
'''ตรวจสอบผลการคำนวณและการแสดงผลลัพธ์ให้เหมือนกับโจทย์ที่กำหนดให้'''

'''Drop 3 days'''
test(
    ['32.6', '34.4', '35.0', '34.6', '34', '35.6', '34'],
    'Temperature dropped on day 4\nTemperature dropped on day 5\nTemperature dropped on day 7\n',
)
test(
    ['43.6', '40.4', '39.0', '45.6', '30.2', '31.6', '35'],
    'Temperature dropped on day 2\nTemperature dropped on day 3\nTemperature dropped on day 5\n',
)
test(
    ['33.6', '34.4', '32.0', '31.5', '37.2', '31.6', '35'],
    'Temperature dropped on day 3\nTemperature dropped on day 4\nTemperature dropped on day 6\n',
)
test(['45', '45', '45', '45', '45', '45', '45'], '')
print('ok')
```
