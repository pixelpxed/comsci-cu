## Q6
จงเขียนโปรแกรมที่รับน้ำหนัก และส่วนสูง จากผู้ใช้ มาคำนวณ bmi โดยให้ระบุหน่วยของน้ำหนักเป็น kg
หรือ lbs และ หน่วยของส่วนสูงเป็น cm หรือ m หรือ ft แล้วไปแปลงเป็น kg และ m และคำนวณแสดงผลลัพธ์ว่าผู้ใช้มีรูปร่างในระดับใด <br>
(1 kg. = 2.205 lbs, 1 m. = 3.2808399 feet)

### ตัวอย่างการทำงาน: 

Weight : **61.3 kg**<br>
Height : **1.65 m**<br>
รูปร่างปกติ

### ตัวอย่างการทำงาน: 

Weight : **159.3 lbs** <br>
Height : **4.92 ft**<br>
อ้วนระดับ 2

### ตัวอย่างการทำงาน:

Weight : **110.8 lbs** <br> Height : **167 cm**<br>ผอม


```python
def run():
    weight_s = input("Weight: ").split(" ")
    height_s = input("Height: ").split(" ")

    # Clean: Get only the numbers. (Split by space, get the front value.)
    weight_c = float(weight_s[0])
    height_c = float(height_s[0])

    # Convert values to used units.
    if weight_s[1] == "lbs":
        weight_c = weight_c / 2.205 # lbs -> kg
    if height_s[1] == "cm":
        height_c = height_c / 100   # cm -> m.
    if height_s[1] == "ft":
        height_c = height_c / 3.2808399 # ft -> m.

    # Calculated with (kg, m)
    bmi = weight_c / (height_c ** 2)

    # Find BMI range descriptions.
    if (bmi < 18.5):
        print("ผอม")
    elif (bmi < 23):
        print("รูปร่างปกติ")
    elif (bmi < 25):
        print("รูปร่างอ้วน")
    elif (bmi < 30):
        print("อ้วนระดับ 1")
    else:
        print("อ้วนระดับ 2")

    pass
```

```python
'''ตรวจสอบผลลัพธ์ของโปรแกรมด้วยการรันที่ cell นี้'''

run()
```

```python
'''ตรวจสอบผลการคำนวณ'''

from unittest.mock import patch
from io import StringIO


def test(inputs, expected_output):
    with patch('builtins.input', side_effect=inputs), patch('sys.stdout', new_callable=StringIO) as mock_stdout:
        run()
        assert mock_stdout.getvalue() == expected_output, mock_stdout.getvalue()
```

```python
'''ตรวจสอบผลการคำนวณและการแสดงผลลัพธ์ให้เหมือนกับโจทย์ที่กำหนดให้'''

test(['61.3 kg', '1.65 m'], 'รูปร่างปกติ\n')  #
test(['61.3 kg', '165 cm'], 'รูปร่างปกติ\n')
test(['49.5 kg', '5 ft'], 'รูปร่างปกติ\n')
test(['110.8 lbs', '150 cm'], 'รูปร่างปกติ\n')
test(['110.8 lbs', '1.50 m'], 'รูปร่างปกติ\n')
test(['92.5 lbs', '4.8 ft'], 'รูปร่างปกติ\n')
print('ok')
```
