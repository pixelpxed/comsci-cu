จงเขียนโปรแกรมที่รับชื่อและอายุจากผู้ใช้ แล้วคำนวณปีเกิดของผู้ใช้เพื่อสร้าง email address ของผู้ใช้โดย
เอาชื่อมาต่อด้วยเลข 2 หลักหลังของปีเกิดและต่อด้วย @chula.ac.th แสดงผลลัพธ์ดังตัวอย่างข้างล่างนี้

### ตัวอย่างการทำงาน:

Name: Somsak <br>
Age : 20 <br>
Somsak47@chula.ac.th

### ตัวอย่างการทำงาน:

### ตัวอย่างการทำงาน:

```python
from datetime import date

def run():
    name = input("Name: ")
    year = str(date.today().year + 543 - int(input("Age: ")))[-2:]

    print(f"{name}{year}@chula.ac.th")
    pass
```

```python
'''ตรวจสอบผลลัพธ์ของโปรแกรมด้วยการรันที่ cell นี้'''
run()
```

```python
from unittest.mock import patch
from io import StringIO
from datetime import date


def test(inputs, expected_output):
    with patch('builtins.input', side_effect=inputs), patch('sys.stdout', new_callable=StringIO) as mock_stdout:
        run()
        assert mock_stdout.getvalue() == expected_output, mock_stdout.getvalue()


def test_easy_print(inputs, expected_output):
    with patch('builtins.input', side_effect=inputs), patch('sys.stdout', new_callable=StringIO) as mock_stdout:
        run()
        assert expected_output in mock_stdout.getvalue(), mock_stdout.getvalue()
```

```python
'''ตรวจสอบการแสดงผลลัพธ์อย่างง่าย'''
test_easy_print(inputs=['Somsak', '65'], expected_output='Somsak')
print('ok')
```

```python
'''ตรวจสอบเลขสองตัวท้ายและการแสดงผลลัพธ์ - กรณี 00-09 (dynamic)'''
from datetime import date
def suffix(age): return str(date.today().year + 543 - age)[-2:]

test(['Somsak','67'], f'Somsak{suffix(67)}@chula.ac.th\n')
test(['Yaya','57'], f'Yaya{suffix(57)}@chula.ac.th\n')
print('ok')
```
