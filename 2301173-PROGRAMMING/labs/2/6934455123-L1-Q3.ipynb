## Q3 
จงเขียนโปรแกรมที่รับชื่อและอายุจากผู้ใช้ แล้วคำนวณปีเกิดของผู้ใช้และแสดงผลลัพธ์ดังตัวอย่างข้างล่างนี้

### ตัวอย่างการรัน: 
*(สมมติให้ปีปัจจุบันคือ 2567)*

Name, age: Somsak 19 <br>
Somsak's year of birth is 2548.

##### ตัวอย่างการรัน:

Name, age: Bella 2 <br>
Bella's year of birth is 2565.

```python
from datetime import date

def run():
    # YOUR CODE HERE
    name, age = input("Name, age: ").split(" ")
    print(f"{name}'s year of birth is {date.today().year + 543 - int(age)}.")
```

```python
'''ตรวจสอบผลลัพธ์ของโปรแกรมด้วยการรันที่ cell นี้'''
run()
```

```python
'''ตรวจสอบการคำนวณปีเกิด'''
from unittest.mock import patch
from io import StringIO
from datetime import date


def test(inputs, expected_output):
    with patch('builtins.input', side_effect=inputs), patch('sys.stdout', new_callable=StringIO) as mock_stdout:
        run()
        assert mock_stdout.getvalue() == expected_output, mock_stdout.getvalue()


def test_easy_print(inputs, expected_output):
    with patch('builtins.input', side_effect=inputs), patch('sys.stdout', new_callable=StringIO) as mock_stdout:
        run()
        assert expected_output in mock_stdout.getvalue(), mock_stdout.getvalue()
```

```python
'''ตรวจสอบการแสดงผลลัพธ์อย่างง่าย'''
test_easy_print(inputs=['Somsak 19'], expected_output="Somsak")
print('ok')
```

```python
'''ตรวจสอบการคำนวณปีเกิดและการแสดงผลลัพธ์ (dynamic)'''
from datetime import date
def by(age): return date.today().year + 543 - age


test(['Somsak 19'], f"Somsak's year of birth is {by(19)}.\n")
test(['Bella 2'],  f"Bella's year of birth is {by(2)}.\n")
print('ok')
```
