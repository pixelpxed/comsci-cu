## Q3
จงเขียนฟังก์ชัน get_dict(file_name) ที่อ่านไฟล์ studentScore.txt ที่มีรายชื่อนักเรียนและคะแนนสอบดังแสดงในตัวอย่างข้างล่างเก็บเป็น dictionary และเขียนฟังก์ชัน rank_students(scores) ที่คืนค่าชื่อนักเรียนเรียงจากคะแนนสูงสุดไปต่ำสุด กำหนดให้ ไม่มีใครได้คะแนนเท่ากัน


### ตัวอย่างไฟล์ studentScore1.txt: 

Piti 25 <br>
Mana 32 <br>
Ari 19 <br>
Suda 28 <br>
Somsak 26 <br>


### ตัวอย่างไฟล์ studentScore2.txt: 

Manee 22 <br>
Piti 32 <br>
Ton 35 <br>
Tree 21 <br>
Siri 23 <br>
Mon 20 <br>


### ตัวอย่างการทำงาน: 
rank_students(get_dict(**'studentScore1.txt'**)) <br>
```['Mana', 'Suda', 'Somsak', 'Piti', 'Ari']```


### ตัวอย่างการทำงาน: 
rank_students(get_dict(**'studentScore2.txt'**)) <br>
```['Ton', 'Piti', 'Siri', 'Manee', 'Tree', 'Mon']```


````python
def get_dict(file_name):
    file = open(file_name, "r")
    lines = file.readlines()

    dictionary = {}
    for line in lines:
        line = line.split()
        dictionary[line[0]] = int(line[1])
        
    file.close()
    return dictionary

def rank_students(scores):
    students = list(scores.keys())
    for i in range(len(students)):
        students[i] = (students[i], scores[students[i]])
    ranked = sorted(students, key = lambda x: x[1], reverse = True)
    return [student[0] for student in ranked]
        
````

````python
'''ตรวจสอบ get_dict'''
assert get_dict('studentScore1.txt') == {
    'Piti': 25,
    'Mana': 32,
    'Ari': 19,
    'Suda': 28,
    'Somsak': 26,
}
assert get_dict('studentScore2.txt') == {
    'Manee': 22,
    'Piti': 32,
    'Ton': 35,
    'Tree': 21,
    'Siri': 23,
    'Mon': 20,
}
print('ok')
````

````python
'''ตรวจสอบ rank_students'''
assert rank_students(get_dict('studentScore1.txt')) == ['Mana', 'Suda', 'Somsak', 'Piti', 'Ari']
assert rank_students(get_dict('studentScore2.txt')) == ['Ton', 'Piti', 'Siri', 'Manee', 'Tree', 'Mon']
print('ok')
````
