# Q5: Library management system

Imagine you are building a simple library management system. Implement three classes: `Author`, `Book`, and `Library`.

The **Author** class should have the following attributes:
- `name (str)`: the name of the author
- `birth_year (int)`: the birth year of the author

It should also have a method:
- `display_info()`: returns the string `Author: {name}, Birth Year: {birth_year}`.


````python
class Author:
    def __init__(self, name: str, birth_year: int):
        self.name = name
        self.birth_year = birth_year

    def display_info(self):
        return f"Author: {self.name}, Birth Year: {self.birth_year}"
````

**Author Class Test Cases:**
- Test Case 1.1: Create an Author instance with a valid name and birth year.
- Test Case 1.2: Verify that display_info() method returns a string.

````python
author = Author("John Doe", 1980)
assert isinstance(author, Author)
assert author.name == "John Doe"
assert author.birth_year == 1980
assert author.display_info() == "Author: John Doe, Birth Year: 1980"
print('ok')
````

The **Book** class should have the following attributes:
- `title (str)`: the title of the book
- `isbn (str)`: the International Standard Book Number of the book
- `author (Author)`: the author of the book (an instance of the Author class)
- `available_copies (int)`: the number of available copies of the book

It should also have the following methods. Each one returns a string. None of them print.

- `display_info()`: returns

```
Title: {title}
ISBN: {isbn}
Author: {author name}, Birth Year: {author birth year}
Available Copies: {available_copies}
```

with a newline after the last line.

- `borrow_copy()`: if `available_copies` is at least 1, decrease it by 1 and return `Book '{title}' borrowed successfully.` If it is 0, leave the count unchanged and return `Sorry, '{title}' is currently unavailable.`
- `return_copy()`: increase `available_copies` by 1 and return `Book '{title}' returned successfully.`


````python
class Book:
    def __init__(self, title: str, isbn: str, author: Author, available_copies: int):
        self.title = title
        self.isbn = isbn
        self.author = author
        self.available_copies = available_copies

    def display_info(self):
        return (
            f"Title: {self.title}\n"
            f"ISBN: {self.isbn}\n"
            f"{self.author.display_info()}\n"
            f"Available Copies: {self.available_copies}\n"
        )

    def borrow_copy(self):
        if (self.available_copies > 0):
            self.available_copies -= 1
            return f"Book '{self.title}' borrowed successfully."
        else:
            return f"Sorry, '{self.title}' is currently unavailable."

    def return_copy(self):
        self.available_copies += 1
        return f"Book '{self.title}' returned successfully."
````

**Book Class Test Cases:**
- Test Case 2.1: Create a Book instance with a valid title, ISBN, Author instance, and available copies.
- Test Case 2.2: Verify that display_info() method returns a string.
- Test Case 2.3: Borrow a copy of a book with available copies.
- Test Case 2.4: Return a copy of a book.
- Test Case 2.5: Attempt to borrow a copy of a book with no available copies.

````python
author = Author("John Doe", 1980)
book = Book("Introduction to Python", "978-1-234567-89-0", author, 10)
assert book.title == "Introduction to Python"
assert book.isbn == "978-1-234567-89-0"
assert book.author is author
assert book.available_copies == 10
assert book.display_info() == (
    "Title: Introduction to Python\n"
    "ISBN: 978-1-234567-89-0\n"
    "Author: John Doe, Birth Year: 1980\n"
    "Available Copies: 10\n"
)
assert book.borrow_copy() == "Book 'Introduction to Python' borrowed successfully."
assert book.available_copies == 9

empty = Book("Advanced Python Programming", "978-1-234567-89-2", author, 1)
assert empty.borrow_copy() == "Book 'Advanced Python Programming' borrowed successfully."
assert empty.available_copies == 0
assert empty.borrow_copy() == "Sorry, 'Advanced Python Programming' is currently unavailable."
assert empty.available_copies == 0

returned = Book("Data Structures in Python", "978-1-234567-89-1", author, 5)
assert returned.return_copy() == "Book 'Data Structures in Python' returned successfully."
assert returned.available_copies == 6
print('ok')
````

The **Library** class should have the following attribute:
- `books`: a dict mapping each book's ISBN to one `Book`. It starts empty.

It should also have the following methods:

- `add_book(book: Book)`: stores the book under its ISBN. If that ISBN is already present, add the new `available_copies` onto the existing book instead of replacing it.
- `display_books()`: returns one string. It starts with `Books in the Library:` and a line of dashes, then each book's `display_info()`, with a blank line between books.
- `search_books_by_author(author_name: str)`: returns a string that starts with `Books by {author_name}:` and a line of dashes, then the matching titles one per line. If none match, the last line is `No books found by {author_name}.`
- `search_books_by_title(title: str)`: if the title is in the library, returns `Book Information for '{title}':`, a line of dashes, then that book's `display_info()`. Otherwise returns `No book found with title '{title}'.`
- `total_available_copies()`: returns the sum of `available_copies` across the books, as an int.

The last cell is a locked example of class composition. You do not write that cell.


````python
class Library:
    def __init__(self):
        self.books = {}

    def add_book(self, book: Book):
        if (book.isbn in self.books):
            self.books[book.isbn].available_copies += book.available_copies
        else:
            self.books[book.isbn] = book

    def display_books(self):
        header = (
            "Books in the Library:\n"
            "---------------------\n"
        )

        details = "\n".join(
            book.display_info()
            for book in self.books.values()
        )
        
        return f"{header}{details}\n"

    def search_books_by_author(self, author_name: str):
        header = (
            f"Books by {author_name}:\n"
            "------------------\n"
        )
        match_author = [
            book.title
            for book in self.books.values()
            if (book.author.name == author_name)
        ]

        if len(match_author) == 0:
            match_author = [f"No books found by {author_name}."]
            
        return header + "\n".join(match_author)

    def search_books_by_title(self, title: str):
        for book in self.books.values():
            if book.title == title:
                header = (
                    f"Book Information for '{title}':\n"
                    "---------------------------------------------\n"
                )
                return header + book.display_info()
        return f"No book found with title '{title}'."
        
    def total_available_copies(self):
        return sum(book.available_copies for book in self.books.values())
````

````python
# Example Usage
author1 = Author("John Doe", 1980)
book1 = Book("Introduction to Python", "978-1-234567-89-0", author1, 10)
book2 = Book("Data Structures in Python", "978-1-234567-89-1", author1, 5)
book3 = Book("Advanced Python Programming", "978-1-234567-89-2", author1, 1)


library = Library()
library.add_book(book1)
library.add_book(book2)
library.add_book(book3)

# Display information about the books in the library
print(library.display_books())

# Borrow a copy of a book
print(library.books[book1.isbn].borrow_copy())

# Return a copy of a book
print(library.books[book2.isbn].return_copy())

# Display updated information about the books in the library
print(library.display_books())

# Search for books by a specific author
print(library.search_books_by_author("John Doe"))

# Search for a book by title
print(library.search_books_by_title("Introduction to Python"))

# Get the total number of available copies in the library
print(library.total_available_copies())
````
