### Q1: Temperature Converter
Write a program that prompts the user for a single integer input representing a temperature value in degrees Celsius ($C$). Your program must convert this value to degrees Fahrenheit ($F$) using the integer conversion formula:

$$F = \left(C \times \frac{9}{5}\right) + 32$$

*Note: Use integer division (`//`) to perform the division.* 

Finally, print the converted temperature matching the label template "Fahrenheit: " followed by the calculated integer value.

**Example Output**
```
Enter Celsius: 25
Fahrenheit: 77
```

````python
def q1():
    cel = int(input("Enter Celsius: "))  # Get user input
    fah = (cel * 9 // 5) + 32            # Calculate to Fahrenheit
    print("Fahrenheit:", fah)            # Returns the result
````

````python
test(['25'], 'Fahrenheit: 77\n', f=q1)
test(['0'], 'Fahrenheit: 32\n', f=q1)
print("ok")
````
