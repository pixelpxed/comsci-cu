## Q4
จงเขียนฟังก์ชัน `quad1(a, b, c)` และ `quad2(a, b, c)` ที่รับค่าสัมประสิทธิ์ $a, b, c$ เป็นจำนวนจริง(จำนวนเต็ม, จำนวนทศนิยม) สำหรับสมการ $ax^2 + bx + c = 0$ ($a \neq 0$):
- `quad1` ส่งคืนค่า $x = \frac{-b + \sqrt{b^2-4ac}}{2a}$ (ปัดเศษทศนิยม 2 ตำแหน่ง)
- `quad2` ส่งคืนค่า $x = \frac{-b - \sqrt{b^2-4ac}}{2a}$ (ปัดเศษทศนิยม 2 ตำแหน่ง)

จากนั้นเขียนฟังก์ชัน `solve_quadratic_file(filename)` ที่อ่านไฟล์สัมประสิทธิ์แล้วคืนค่า list ของคำตอบแต่ละบรรทัด:
- คืนค่า tuple `(x1, x2)` หากหาคำตอบได้
- คืนค่า string `"Invalid problem"` หากหาคำตอบไม่ได้ ($a = 0$ หรือ $b^2-4ac < 0$)

### ตัวอย่างการทำงาน:

```python
quad1(3, -7, 2) # คืนค่า 2.0
quad2(3, -7, 2) # คืนค่า 0.33

solve_quadratic_file('problems1.txt')
# คืนค่า [(1.0, 1.0), (2.0, 0.33), (1.0, 0.4), (-0.26, -1.54)]

solve_quadratic_file('problems2.txt')
# คืนค่า ['Invalid problem', (-1.0, -3.0), (1.0, 0.4), 'Invalid problem']
```

````python
import math

def quad1(a, b, c): # Plus
    return round((((-1 * b) + math.sqrt((b ** 2) - (4 * a * c))) / (2 * a)), 2)

def quad2(a, b, c): # Minus
    return round((((-1 * b) - math.sqrt((b ** 2) - (4 * a * c))) / (2 * a)), 2)

def solve_quadratic_file(filename):
    file_raw = open(filename)
    file_contents = file_raw.read().split("\n")

    answers = []

    # Clean empty strings out of data set.
    while ("" in file_contents):
        file_contents.remove("")

    for i in range(len(file_contents)):
        a, b, c = file_contents[i].split(" ")

        a = float(a)
        b = float(b)
        c = float(c)

        if (
            a == 0 or
            ((b ** 2) - (4 * a * c)) < 0
        ):
            answers.append("Invalid problem")
        else:
            answers.append((quad1(a, b, c), quad2(a, b, c)))

    file_raw.close()
    return answers
````

````python
'''ตรวจสอบฟังก์ชัน quad1'''
assert quad1(1, -2, 1) == 1.0
assert quad1(3, -7, 2) == 2.0
assert quad1(5, 9, 2) == -0.26
print('ok')
````

````python
'''ตรวจสอบฟังก์ชัน quad2'''
assert quad2(1, -2, 1) == 1.0
assert quad2(3, -7, 2) == 0.33
assert quad2(5, 9, 2) == -1.54
print('ok')
````

````python
'''ตรวจสอบฟังก์ชัน solve_quadratic_file'''
assert solve_quadratic_file('problems1.txt') == [
    (1.0, 1.0),
    (2.0, 0.33),
    (1.0, 0.4),
    (-0.26, -1.54)
]

assert solve_quadratic_file('problems2.txt') == [
    'Invalid problem',
    (-1.0, -3.0),
    (1.0, 0.4),
    'Invalid problem'
]
print('ok')
````
