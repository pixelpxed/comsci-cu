## Q2
จงเขียนฟังก์ชัน `move_robot_with_wall(filename, initial_pos)` ที่รับชื่อไฟล์คำสั่ง `filename` และตำแหน่งเริ่มต้น `initial_pos` ในรูปแบบ tuple `(x, y)`
ให้ อ่านลำดับคำสั่งเคลื่อนที่หุ่นยนต์ไปทางซ้าย (L), ขวา (R), ขึ้นบน (U), ลงล่าง (D) ครั้งละ 1 หน่วย จากไฟล์
โดยมีกำแพงที่เส้น `x=10, x=-10, y=10 และ y=-10` (หุ่นยนต์สามารถเดินได้ในช่วง `-9 <= x <= 9` และ `-9 <= y <= 9` หากเดินชนกำแพง จะไม่เคลื่อนที่)

การคืนค่า (return):
- คืนค่าตำแหน่งสุดท้ายเป็น tuple `(x, y)` หากคำสั่งทั้งหมดถูกต้อง
- คืนค่า `"Invalid command"` หากพบคำสั่งที่ไม่ถูกต้อง

### ตัวอย่างการทำงาน:

```python
move_robot_with_wall('move3.txt', (9, -9)) # คืนค่า (9, -9)
move_robot_with_wall('move3.txt', (8, 4))  # คืนค่า (9, 3)
move_robot_with_wall('move4.txt', (1, 2))  # คืนค่า "Invalid command"
```

````python
def move_robot_with_wall(filename, initial_pos):
    file_raw = open(filename)
    file_contents = file_raw.read().split("\n")

    pos = initial_pos

    for i in range(len(file_contents)):
        if (file_contents[i] not in ["L", "R", "U", "D"]):
            return "Invalid command"

        if (
            file_contents[i] == "L" and
            pos[0] - 1 > -10
        ):
            pos = (pos[0] - 1, pos[1])
        elif (
            file_contents[i] == "R" and
            pos[0] + 1 < 10
        ):
            pos = (pos[0] + 1, pos[1])
        elif (
            file_contents[i] == "U" and
            pos[1] + 1 < 10
        ):
            pos = (pos[0], pos[1] + 1)
        elif (
            file_contents[i] == "D" and
            pos[1] - 1 > -10
        ):
            pos = (pos[0], pos[1] - 1)

    file_raw.close()
    return pos
````

````python
'''ตรวจสอบการคำนวณตำแหน่งหุ่นยนต์ร่วมกับเงื่อนไขกำแพง'''
assert move_robot_with_wall('move3.txt', (9, -9)) == (9, -9)
assert move_robot_with_wall('move3.txt', (8, 4)) == (9, 3)
assert move_robot_with_wall('move3.txt', (9, 5)) == (9, 4)
assert move_robot_with_wall('move3.txt', (-5, 5)) == (-3, 4)
assert move_robot_with_wall('move3.txt', (-1, -1)) == (1, -2)
print('ok')
````

````python
'''ตรวจสอบคำสั่งที่ไม่ถูกต้อง (Invalid command)'''
assert move_robot_with_wall('move4.txt', (15, 20)) == "Invalid command"
assert move_robot_with_wall('move4.txt', (1, 1)) == "Invalid command"
assert move_robot_with_wall('move4.txt', (1, 2)) == "Invalid command"
print('ok')
````
