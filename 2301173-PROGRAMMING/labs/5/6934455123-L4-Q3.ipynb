## Q3
จงเขียนฟังก์ชัน `f(x)` ที่คำนวณ $1/x!$ โดยใช้ฟังก์ชัน `factorial` จาก module `math` (ปัดเศษทศนิยม 2 ตำแหน่ง)
และเขียนฟังก์ชัน `summation(m, n)` ที่คืนค่าผลรวมของ `f(x)` เมื่อ `x` มีค่าตั้งแต่ `m` ถึง `n`
โดยหาก `m > n` ให้สลับค่าเพื่อคำนวณจากค่าน้อยไปค่ามากเสมอ

**การคืนค่า:** คืนค่าผลลัพธ์ในรูปทศนิยมที่ปัดเศษเหลือ 2 ตำแหน่ง โดยใช้ฟังก์ชัน `round`

### ตัวอย่างการทำงาน:

```python
f(3)            # คืนค่า 0.17
summation(1, 3) # คืนค่า 1.67
summation(3, 1) # คืนค่า 1.67
summation(3, 6) # คืนค่า 0.22
```

````python
import math

def f(x):
    return round(1 / math.factorial(x), 2)

def summation(m, n):
    if (m > n): m, n = n, m # Always least -> most.

    sum = 0
    for i in range(m, n + 1):
        sum = sum + f(i)

    return round(sum, 2)
````

````python
'''ตรวจสอบฟังก์ชัน f(x)'''
assert f(1) == 1.0
assert f(2) == 0.5
assert f(3) == 0.17
assert f(4) == 0.04
print('ok')
````

````python
'''ตรวจสอบฟังก์ชัน summation(m, n) กรณี m <= n'''
assert summation(1, 3) == 1.67
assert summation(3, 6) == 0.22
assert summation(5, 6) == 0.01
assert summation(3, 4) == 0.21
print('ok')
````

````python
'''ตรวจสอบฟังก์ชัน summation(m, n) กรณี m > n'''
assert summation(3, 1) == 1.67
assert summation(9, 2) == 0.72
assert summation(5, 2) == 0.72
print('ok')
````
