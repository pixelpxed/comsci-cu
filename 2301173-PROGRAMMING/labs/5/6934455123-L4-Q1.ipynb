## Q1
จงเขียนฟังก์ชัน `move_robot(filename, initial_pos)` ที่รับชื่อไฟล์คำสั่ง `filename` และตำแหน่งเริ่มต้น `initial_pos` ในรูปแบบ tuple `(x, y)`
ให้ อ่านลำดับคำสั่งเคลื่อนที่หุ่นยนต์ไปทางซ้าย (L), ขวา (R), ขึ้นบน (U), ลงล่าง (D) ครั้งละ 1 หน่วย จากไฟล์ แล้วคืนค่า (return) ดังนี้:
- คืนค่าตำแหน่งสุดท้ายเป็น tuple `(x, y)` หากคำสั่งทั้งหมดถูกต้อง
- คืนค่า `"Invalid command"` หากพบคำสั่งที่ไม่ถูกต้อง

### ตัวอย่างการทำงาน:

```python
move_robot('move1.txt', (10, 20)) # คืนค่า (8, 23)
move_robot('move2.txt', (10, 20)) # คืนค่า "Invalid command"
```

````python
def move_robot(filename, initial_pos):
    file_raw = open(filename)
    file_contents = file_raw.read().split("\n")

    pos = initial_pos

    for i in range(len(file_contents)):
        if (file_contents[i] == "L"):
            pos = (pos[0] - 1, pos[1])
        elif (file_contents[i] == "R"):
            pos = (pos[0] + 1, pos[1])
        elif (file_contents[i] == "U"):
            pos = (pos[0], pos[1] + 1)
        elif (file_contents[i] == "D"):
            pos = (pos[0], pos[1] - 1)
        else:
            return "Invalid command"

    file_raw.close()
    return pos
````

````python
'''ตรวจสอบการคำนวณตำแหน่งหุ่นยนต์'''
assert move_robot('move1.txt', (10, 20)) == (8, 23)
assert move_robot('move1.txt', (7, 5)) == (5, 8)
assert move_robot('move1.txt', (100, -20)) == (98, -17)
assert move_robot('move1.txt', (-100, 50)) == (-102, 53)
assert move_robot('move1.txt', (-100, -100)) == (-102, -97)
print('ok')
````

````python
'''ตรวจสอบคำสั่งที่ไม่ถูกต้อง (Invalid command)'''
assert move_robot('move2.txt', (15, 20)) == "Invalid command"
assert move_robot('move2.txt', (1, 1)) == "Invalid command"
print('ok')
````
