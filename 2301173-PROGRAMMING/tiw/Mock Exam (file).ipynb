## Exercise: ตัดสินผลการประกวดร้องเพลง

จงเขียนโปรแกรมเพื่อตัดสินผลการประกวดร้องเพลง โดยให้โปรแกรมอ่านคะแนนที่ผู้เข้าประกวดได้รับจากคณะกรรมการ (เต็ม 10 คะแนนสำหรับกรรมการแต่ละท่าน) และผู้ชมในห้องส่ง (เต็ม 70 คะแนน) จากไฟล์ `score<n>.txt` โดยแต่ละบรรทัดประกอบด้วย รหัสผู้ประกวด, คะแนนกรรมการคนที่ 1, คะแนนกรรมการคนที่ 2, คะแนนกรรมการคนที่ 3 และคะแนนผู้ชมในห้องส่ง คั่นด้วยช่องว่าง

กำหนดให้เขียนฟังก์ชัน `read_scores(file_name)` ที่รับชื่อไฟล์เป็น parameter แล้ว คำนวณคะแนนรวมของผู้เข้าประกวดแต่ละคน แล้วคืนค่าเป็น **list ของ tuple** โดยแต่ละ tuple มีรูปแบบ `(รหัสผู้เข้าประกวด, คะแนนรวม)` เรียงตามลำดับที่พบในไฟล์

จงเขียนฟังก์ชัน `find_top3(scores)` ที่รับ parameter เป็น list ของ tuple ที่ได้จาก `read_scores` แล้ว เปรียบเทียบคะแนนเพื่อคืนค่าเป็น list ของรหัสผู้มีคะแนนสูงสุด 3 อันดับแรก (กำหนดให้ไม่มีผู้ใดได้คะแนนเท่ากัน)

1. read file
2. .split()
3. list -> (name, sum(scores))
4. sort scores max->min
5. return first 3 arrays

```python
def read_scores(file_name):
    # Reading file
    file_raw = open(file_name)
    file_content = file_raw.readlines()
    file_raw.close()

    scores = []

    for score in file_content:
        score = score.split()
        scores.append(
            (
                score[0],
                sum(
                    [
                        int(score[1]),
                        int(score[2]),
                        int(score[3]),
                        int(score[4])
                    ]
                )
            )
        )

    return scores

def getTheScore(score_tuple):
    return score_tuple[1]

def find_top3(scores):
    # scores_sorted = sorted(scores, key=getTheScore, reverse=True)
    scores_sorted = sorted(scores, key = lambda x:x[1], reverse=True)

    return [
        scores_sorted[0][0],
        scores_sorted[1][0],
        scores_sorted[2][0]
    ]




def run():
    filename = input("Enter file name: ")
    result = find_top3(read_scores(filename))
    print(f"The winners are :\n{result}")
```

```python
'''ตรวจสอบการสร้าง list ของ tuple'''
'''ตัวอย่าง read_scores("ใส่ชื่อไฟล์.txt")'''
read_scores('score1.txt')
```

```python
'''ทดสอบฟังก์ชัน read_scores(file_name)'''
result1 = read_scores('score1.txt')
assert result1 == [('CU01', 90), ('CU02', 63), ('CU03', 78), ('CU04', 77), ('CU05', 93)]

result2 = read_scores('score2.txt')
assert result2 == [('CU01', 90), ('CU02', 63), ('CU03', 77), ('CU04', 77), ('CU05', 93), ('CU06', 86), ('CU07', 94)]

result3 = read_scores('score3.txt')
assert result3 == [('CU01', 90), ('CU02', 63), ('CU03', 77), ('CU04', 77), ('CU05', 87), ('CU06', 86), ('CU07', 94), ('CU08', 84), ('CU09', 79), ('CU10', 86)]

print('OK')
```

```python
'''ทดสอบฟังก์ชัน find_top3()'''
'''จากตัวอย่าง list ของ tuple ผู้มีคะแนนสูงสุด 3 ลำดับแรกคือ E, B และ F ตามลำดับ'''
assert find_top3([('A',80), ('B',90), ('C',70), ('D',50), ('E',95), ('F',86)]) == ['E', 'B', 'F']
assert find_top3([('A',97), ('B',68), ('C',96), ('D',88), ('E',85), ('F',86)]) == ['A', 'C', 'D']
print('OK')
```
