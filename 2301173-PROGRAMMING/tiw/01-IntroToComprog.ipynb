## Python Basics

### `print()` : Displays output.
- By default, Python automatically includes `\n` in the end of each print.
  - `\n` represents a new line, you can type this inside a string to break content between lines.
  - To override this, put `end=""` as the second argument inside `print()`.

Here are some examples:


````python
# Displaying data:
print("Hello, world!")    # Displays "Hello, world!\n"
print(12345)              # Displays "12345\n"
````

````python
# Returning nothing:
print("=== (Start) ===")
print("")                 # Returns "\n" (new line, displayed as empty line)
print("--- (Break) ---")
print()                   # Returns "\n" (new line, displayed as empty line)
print("--- (Break) ---")
print("", end="")         # Returns ""   (Empty string, same as nothing inside output window.)
print("--- (Break) ---")
print("===  (End)  ===")

# You can see the gap between start, 1st and 2nd (Break), as there's a empty line betweens it.
````

````python
# Using `end="{string}`
print("Hello, ", end="")  # Displays "Hello, "
print("world!")           # Displays "world!"
# Combined together, it displays "Hello, world!"
````

````python
# A function that returns something.
def pythonReturnsStr():
  return "Something!"
def pythonReturnsInt():
  return 123123

# Printing a value returned from a function.
print(pythonReturnsStr())
print(pythonReturnsInt())
````

### Data Types

Data types are how data/value is handled in Python, mainly there's:
- `str` a string value. ()
  - Texts, that should not be calculatable.
    ex. `Hello, world!`, `Something123`, `Imagine a place.`

- `int` an integer value. (eg. 1, 2, 3, 4, 5)
  - Use this when handling things that should be countable by hand.  
    ex. `for i in range(5):`, `while i <= 10:`, etc.

- `float` a decimal value. (eg. 1.0, 1.1, 1.01)
  - Use this when you need exact calculations of something.
    ex. `22.5 * 3.04`
  - Note: When you want an integer to be a float, add `.0` after value. (ex. `1.0`)

Using the right values will help your program be more precise and stable.

````python
print("--- (String)  ---")

# String
print(type("Hello, world!"))
print(type("Something123"))
print(type("Imagine a place."))

print("\n--- (Integer) ---")

# Integer
print(type(1))
print(type(2))
print(type(3))

print("\n---  (Float)  ---")

print(type(3.0))
print(type(3.14))
print(type(3.141))

# Float
````

## Lists, Tuples, and Dictionaries
Putting this before we go further might make you understand better, so let's go 
over some data types that you can use to make your life easier.

- Lists: A type of data that's in a list.  
  For example: Putting your name inside a 1. 2. 3. list to go somewhere.
- Tuples: Great for ensuring your data length or consistency doesn't break.  
  For example: (x, y) positions
- Dictionaries: A table you can lookup on for data.
  For example: A price for items

````python
# Lists

lists = [1, 2, 3, 4, 5]
# pos:   0, 1, 2, 3, 4, 5, ...
# Python starts counting at 0, then increments by 1
print(type(lists))
print(lists[0], lists[2], lists[4])
print(lists)

print("\n--- (Break) ---\n")

# Adding, editing a value to list.
lists[0] = 8       # Editing a value in a position
lists.append(6)    # Adding the value to last in list
lists.insert(2, 7) # Adding the value inside the specific position
lists.remove(5)    # Removes the first-matching value inside list

print(lists)
````

````python
# Tuples
tuples = (12, 13)
print(type(tuples))
print(tuples[1], tuples[0])
print(tuples)

print("\n--- (Break) ---\n")

# Editing a value to dictionary
# Because tuples are strict, unlike lists you'll have to build the whole () again.
tuples = (14, tuples[1])
print(tuples)
````

````python
# Dictionaries
dictionary = {
  "key1": "value1",
  "key2": "value2"
}

type(dictionary)
dictionary["key1"]
dictionary.keys()                 # Return all keys.
list(dictionary.keys())           # Return all keys in an list.
list(dictionary.items())          # Return all keys, values as tuples in an list.

# Adding, editing a value to dictionary
dictionary["key2"] = "value4"     # Edits a value
dictionary["something"] = "that"  # Adds a new value
````

## Selections

`if`, `else`, and `elif` are the staple of selecting data only when the conditions match.  
For a real world example, you might only top up fuel if the gas price is under `40 THB`.

While each one of it explains itself, you can remember each as:
- `if` : `if (this) is true, do (this).`
- `else` : `if (this) is true, do (this).` `but if not, do (that)`
- `elif` : `If the above (if) is not true,` `but (this) is true, do (this).`

Note: `elif` means `else if` in Python.

````python
# `if` statements

print("--- (True)  ---")

# True
print(1 == 1)
if (True):
  print("This will show up in output")
if (1 == 1):
  print("This will show up in output")

print("\n--- (False) ---")

# False
print(1 == 2)
if (False):
  print("This will *not* show up in output")
if (1 == 2):
  print("This will *not* show up in output")
````

````python
# `if, else` statements

# True
if (1 == 1):
  print("True")
else:
  print("False")

# False
if (1 == 2):
  print("True")
else:
  print("False")
````

````python
# `if, elif, else` statements

# Let's say we give a value as `5`
value = 2

# This will return "Five".
if (value <= 3):      # Does the `value` more than or equals `1`? No, Skip to the next condition.
  print("Less than or equals three")
elif (value <= 2):    # Is the above condition `True`? No, does the `value` more than or equals  `2`? 
                      # Yes. Run the following instruction(s).
    print("Less than or equals two") # Display "Two"
elif (value <= 1):    # Is the above condition `True`? Yes, no need to check further.
  print("Less than or equals one")
else:                 # Is *all of* the above condition `True`? 
                      # Yes, no need to run the following instruction(s).
  print("None of these statements match.")

print("--- (Break) ---")

# If you need multiple outputs when conditions are met. Just use `if`.

if (value <= 3):      # Does the `value` more than or equals `1`? No, Skip to the next condition.
  print("Less than or equals three")
if (value <= 2):
    print("Less than or equals two")
if (value <= 1):
  print("Less than or equals one")
if (value <= 0):
  print("None of these statements match.")


````

## Repetitions

Repititions are when you need a program to do something, either exactly or similarly. 
For example, counting from 1-100, or looping through a list.  

There's multiple methods to choose from, depending on your preferences and feasability.
- `for i in range(n)`
- `for data in datas`
- `while (True)`

Here are some examples to use:

````python
# for i in range(n)

for i in range(5):
  print(f"Count {i}")

print("\n---    (start, end)     ---")

for j in range(1, 5):
  print(j)

print("\n--- (start, end, steps) ---")

for l in range(0, 25, 5):
  print(l)

print("\n------\n")

# In the negatives
for k in range(5, 0, -1):
  print(k)

print("\n------\n")

# In the negatives, multiple steps
for m in range(25, 0, -5):
  print(m)
````

````python
# for item in items

# First, we need an example list.
items = ["Apple", "Banana", "Coconut"]

print(f"Contents within list: {items}")
for item in items:
  print(item)

print("--- (Adding more items to list) ---")

items.append("Dragonfruit") # .append(item)
print(f"Contents within list: {items}")
for item_2 in items:
  print(item_2)
````

````python
# while (True)

i = 1
while (i <= 10):
  print(i)
  i = i + 1 # Or you could do `i += 1`

array = [1, 1, 1, 2, 3, 5, 6]
while (1 in array):
  array.remove(1)
print(array)
````

## Files and Functions

**Files**: Python can easily opens a file with a single command 
`file = open(filename)` if you put it into the same directory (Means putting it 
in the same folder). When opened, it consumes your memory (RAM). So don't forget 
to `file.close()` when you're done with it.

Related functions:
- `.split()` splits the data inside (commonly: `,`, ` `, `\n`)
- Removing blank lines from your data  
  ```while ("\n" in file_contents.split("\n")): file_contents.remove("\n")```

**Functions**: are a basis of *readable* codes, it's useful when your code needs 
to do a specific thing multiple times. It's a instruction set that when called 
does the instuctions inside: processing, calling other functions, and returning 
data. `def` means `define`, and functions can accept multiple arguments, 
separated by a comma.
```
def functionName(parameter, parameter2):
  # Your code inside
```
To `return` a function, means the rest of the functions below wouldn't run if the
conditions are met already.

````python
# Opening a file

file_raw = open("file_name.txt")
file_contents = file_raw.read()
file_contents = file_raw.readlines()
file_raw.close()

file_contents.split("\n")
file_contents.split(" ")
file_contents.split(", ")
file_contents.split(",")
file_contents.split()
````

````python
# Functions

def functionName(parameter):
  return f"Parameter: {parameter}"
def addTwo(one, two):
  result = one + two
  return f"Adding values: {result}"

print(functionName(6934455123))
print(functionName("pixel.pxed@gmail.com"))
print(addTwo(12, 13))
````

## Related Functions (Good to know)

````python
# Converting value types

print(int(1.00)) # Float -> Integer
print(int(1.99)) # Float -> Integer 
                 # ignores the decimal values and return the integers only.
print(int("12")) # String -> Integer
print(int("12")) # String -> Integer

print(float(123))    # Integer -> Float (123 -> 123.0)
print(float(123.00)) # Float -> Float (123.00 -> 123.00)

print(str(1234)) # Integer -> String

# * Use int() when the value should be counted as 1, 2, 3, ...
# * Use float() when the value should be precise. (ex. 3.141)
# * Use str() when the contents should be a text, non-calcuated with math.

print("\n--- (round()) ---\n")

# Rounding a value
# round(value, digits to round to)
print(round(12.34567, 1)) # 12.3
print(round(12.34567, 2)) # 12.35
print(round(12.34567, 3)) # 12.346

print("\n--- (.strip()) ---\n")

# Cleaning a string's front and back.
# .strip()
print(" something ".strip())
print(" something".strip())
print("something ".strip())

print("\n--- (max, min, sum, avg) ---\n")

array = [1, 2, 3, 4, 5]
# Getting the count of elements in a list.
print(len(array))
# Getting the max, min, sum, avg value from the list.
print(max(array))
print(min(array))
print(sum(array))
print(sum(array) / len(array))

print("\n--- (Selecting datas) ---\n")

# Getting the specific string from specific location.
string = "Something123"

print(string)
print(string[2])    # Getting the 3rd character only.
print(string[-2])    # Getting the 3rd character only.

print(string[:2])   # Gets the first 2 characters
print(string[4:])   # Removes the first 5 characters

print(string[-2:])  # Gets the last 2 characters
print(string[:-4])  # Removes the last 5 characters

print(string[2:5])  # Gets the 2-4 value from string
                    # [index(from), index(to + 1)]
````

## Extra Tips

- Easier way to do things (in my opinion)
  - F-String to format your output
- Debugging your error
  - Help! My tests isn't passing.

````python
# F-Strings are a method to display an output inside just one string,
# no need to seperate methods with commas, plus signs, and work with formatting.
#
# To use a f-string, put f before any stirng "", then to put a variable or a 
# logic inside it, put a curly bracket inside the string. {}

# For example:
name, balance = "Schengen", 100
print(name + "'s Balance: " + str(balance) + " THB")  # Plus method: You need to convert int to string within combining strings.
print((name + "'s"), "Balance:", balance, "THB")      # Comma method
print(f"{name}'s Balance: {balance} THB")             # F-String

# Logic inside brackets:
print(name, "is unfortunate:", balance < 1000)        # Comma method
print(f"{name} is unfortunate: {balance < 1000}")     # F-String

# You can see the output is the same, but our work and readability is much better.
````

````python
# sorted()

# เอาค่าใน tuple แค่ตัวที่ 2 (จุดที่ 1) - เช่น ('CU01', 93) คืนค่า 93
def get_second_value(score_tuple):
  return score_tuple[1]

# sorted() in here!!
def sort_values(scores):
  scores_sorted = sorted(scores, key=get_second_value, reverse=True)
  return scores_sorted

# sorted(
#   list ที่จะ sort, 
#   key=ค่าที่จะใช้ ส่งเป็น function ไม่ต้องใส่ ()
#   reverse=[True หรือ False] ย้อนค่าจาก น้อย -> มาก เป็น มาก -> น้อย มั้ย
# )

sort_values([('CU01', 90), ('CU02', 63), ('CU03', 78), ('CU04', 77), ('CU05', 93)])
````
