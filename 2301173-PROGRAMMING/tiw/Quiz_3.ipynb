## Quiz 3: นับสระและพยัญชนะในข้อความ (Count Vowels and Consonants)

จงเขียนฟังก์ชัน `count_vowels_and_consonants(text)` ที่รับสตริง `text` 
แล้วใช้ `for loop` ในการตรวจสอบอักขระแต่ละตัวเพื่อนับ:
1. **สระภาษาอังกฤษ (Vowels):** ได้แก่ A, E, I, O, U (ไม่สนใจตัวพิมพ์เล็กหรือพิมพ์ใหญ่)
2. **พยัญชนะภาษาอังกฤษ (Consonants):** ตัวอักษรภาษาอังกฤษ A-Z อื่นๆ ที่ไม่ใช่สระ

**ข้อกำหนด:**
- อักขระที่ไม่ใช่ตัวอักษรภาษาอังกฤษ เช่น ตัวเลข, เครื่องหมายวรรคตอน, เว้นวรรค **ไม่ต้องนับ**
- คืนค่า (return) เป็น `tuple` ในรูปแบบ `(vowels, consonants)`

### ตัวอย่างการทำงาน:
```python
count_vowels_and_consonants("Hello, World!")  # คืนค่า (3, 7)
count_vowels_and_consonants("Python 3.12")     # คืนค่า (1, 5)
count_vowels_and_consonants("")               # คืนค่า (0, 0)
count_vowels_and_consonants("AEIOU aeiou")     # คืนค่า (10, 0)
```


````python
def count_vowels_and_consonants(text):
    text = text.upper()

    count_vowels = 0
    count_nonvowels = 0

    for i in range(len(text)):
        if (text[i] in "AEIOU"):
            count_vowels += 1
        if (text[i] in "BCDFGHJKLMNPQRSTVWXYZ"):
            count_nonvowels += 1

    return (count_vowels, count_nonvowels)

count_vowels_and_consonants("Hello, world!")
````

````python
def count_vowels_and_consonants(text):
    text = text.upper()

    count_vowels = 0
    count_nonvowels = 0

    for i in range(len(text)):
        if (text[i].isalpha()):
            if (text[i] in "AEIOU"):
                count_vowels += 1
            else:
                count_nonvowels += 1

    return (count_vowels, count_nonvowels)

count_vowels_and_consonants("Hello, world!")
````

````python
# ตรวจสอบความถูกต้องของผลลัพธ์
assert count_vowels_and_consonants("Hello, world!") == (3, 7), f"Failed: {count_vowels_and_consonants('Hello, World!')}"
assert count_vowels_and_consonants("Python 3.12") == (1, 5), f"Failed: {count_vowels_and_consonants('Python 3.12')}"
assert count_vowels_and_consonants("") == (0, 0), f"Failed: {count_vowels_and_consonants('')}"
assert count_vowels_and_consonants("AEIOU aeiou") == (10, 0), f"Failed: {count_vowels_and_consonants('AEIOU aeiou')}"
assert count_vowels_and_consonants("12345 @#$%") == (0, 0), f"Failed: {count_vowels_and_consonants('12345 @#$%')}"
print("All test cases passed!")

````
