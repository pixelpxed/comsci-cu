Q1: ให้รับรัศมีของวงกลมและความยาวด้านของสี่เหลี่ยมจัตุรัสจากผู้ใช้ แล้วแสดงผลลัพธ์ว่า True ถ้าวงกลมมีพื้นที่มากกว่าสี่เหลี่ยมจัตุรัส แสดงผลลัพธ์ว่า False ถ้าวงกลมมีพื้นที่ไม่มากกว่าสี่เหลี่ยมจัตุรัส (ใช้ ค่าคงที่ pi จาก math library)

```html
Enter radius and length of side: 2.5, 4.1<enter>
This circle is larger than this square: True
```

````python
import math

# Get required values.
radius, length = input("Enter radius and length of side: ").split(",")

# Calculate area of both elements, then compare.
condition = ((math.pi * float(radius)) ** 2) > (float(length) ** 2)

# Return final product.
print(f"This circle is larger than this square: {condition}")
````

Q2: ให้รับชื่อและระยะเวลาที่เป็นผู้ใช้ผลิตภัณฑ์เป็นเดือน แล้วแสดงผลลัพธ์ว่า ผู้ใช้นั้น ใช้ผลิตภัณฑ์เป็นเวลากี่ปีและกี่เดือน

```html
Enter name and no. of months: Tom 42<enter>
Tom has used this product for 3 year 6 month.
```

````python
# Get required values.
name, month = input("Enter name and no. of months: ").split(" ")

# Calculate month and years.
years = int(month) // 12
month_left = int(month) - (years * 12)

# Extra: Plural wording considerations for number counts.
year_wformat = "years" if (years != 1) else "year"
month_wformat = "months" if (month_left != 1) else "month"

# Return final product.
print(f"{name} has used this product for {years} {year_wformat} {month_left} {month_wformat}.")
````

Q5. Write a program that receives three numbers `a`, `b`, `c` separated by `,` in a single line, calculate using this formula
$-b \pm \dfrac{\sqrt{b^2-4ac}}{2a}$, then print out the solution(s) on the screen. Print out only one value if there is a single solution. Print the message "No real solution" if the value within the square root is negative.

````python
import math

a, b, c = input("Variables a, b, c: ").split(",")

a = float(a)
b = float(b)
c = float(c)

rooted_parts = (b ** 2) - (4 * a * c)
                         
if (rooted_parts < 0 or a == 0):
  print("No real solutions.")
else:
  positive = (-1 * b) + math.sqrt((rooted_parts) / (2 * a))
  negative = (-1 * b) - math.sqrt((rooted_parts) / (2 * a))

  if (positive == negative):
    print(f"One solution: {positive}.")
  else: 
    print(f"Two Solutions: {positive}, {negative}.")
````

Q6. Write a program that recieves weight in kg and height in cm, calculate Bio Mass Index; $BMI = w/h^2$ (w in kg and h in m), and print out the corresponding BMI stage  shown below.

**BMI** | $<17.0$ | $<25.0$ | $<40.0$ | $\geq 40.0$
----|-------|-------|-------|------
**Stage** | Underweight | Normal | Overweight ​| Obese

Ex.
```
Weight (kg) = 72<enter>
Height (cm) = 169<enter>
Overweight
```

```
Weight (kg) = 54<enter>
Height (cm) =160<enter>
Normal
```

````python
weight = int(input("Weight (kg): "))
height = int(input("Height (cm): "))

bmi = weight / ((height / 100) ** 2)

if (bmi < 17):
  print(f"Underweight (BMI: {bmi})")
elif (bmi < 25):
  print(f"Normal (BMI: {bmi})")
elif (bmi < 40):
  print(f"Overweight (BMI: {bmi})")
else:
  print(f"Obese (BMI: {bmi})")
````

Q8. Write a program that receives a month and a year in AD and print out the number of days of the month in that year. Note that in the leap year, there are 29 days in Feb - it is when the year in AD is multiple of 4 but not 100 or multiple of 400.

Ex.
```
Enter month: 2<enter>
Enter year: 2009<enter>
Number of days: 28
```
```
Enter month: 2<enter>
Enter year: 2008<enter>
Number of days: 29
```
```
Enter month: 3<enter>
Enter year: 2016<enter>
Number of days: 31
```

````python
month = int(input("Enter month: ")) - 1  # -1 to consult with `data` array.
year = int(input("Enter year: "))

# Not accounting for February leap year!
data = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

# Get days in month from the `data` array.
days = data[month]

# Accounting for leap year edge cases.
if (month == 1) and (  # If February.
  ((year % 4 == 0) and (year % 100 != 0)) or 
  (year % 400 == 0)
):
  days = 29

print(f"Number of days: {days} (Year: {year})")
````

Q10. **Tax calculator**

Write a program that recieves the user's firstname, lastname, salary, and extra annual allowance, then print out the tax amount they have to pay based on the criteria table below.

Net income | Tax rate
-------------| ------
0 - 150,000  | Waived
150,000 - 300,000 | 5%
300,000 - 500,000 | 10%
500,000 | 15%



````python
name = input("Name: ")
salary = float(input("Salary (Monthly): "))
extra = float(input("Extra annual allowance: "))

# Get net income (Escape decimal miscaluclations)
net_income = (((salary * 100) * 12) / 100) + extra

# For calculations
process_income = net_income
income_taxed = 0

rates = [[500000, 0.15], [300000, 0.10], [150000, 0.05]]
for i in range(len(rates)):
  if (process_income > rates[i][0]):
    income_taxed = income_taxed + ((process_income - rates[i][0]) * rates[i][1])
    process_income = process_income - (process_income - rates[i][0])

print(f"As {name}'s net income is '{net_income} THB', they would be taxed '{income_taxed} THB'.")
````
