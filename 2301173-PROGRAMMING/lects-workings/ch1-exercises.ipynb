### Helper Functions
**Run these before you run any other modules.**
Theses functions are to help with repetitive instructions 
in more than one assignments.

````python
# Returns a grade from A-F, based on regular grading conventions.
def getGradeLetter(score):
    if score >= 80:
        return "A"
    if score >= 70:
        return "B"
    if score >= 60:
        return "C"
    if score >= 50:
        return "D"
    return "F"
````

Q4. Write a program that receives three integers in a single line separated by a comma `,` and print out the summation on the screen as shown below.
```
<1stInteger> + <2ndInteger> + <3rdInteger> = <summation>
```

````python
# In Class Method: Seperates by `Commas`.

int1, int2, int3 = input("Value 1, 2, 3 [ex. '1, 2, 3']: ").split(", ")

# Can also be put in the print function;
# this one is to prittify and escape 80 characters limit.
result = int(int1) + int(int2) + int(int3)

print(f"{int1} + {int3} + {int3} = {result}")
````

````python
# Own Method: Seperates by `Array`.

intSets = input("Value 1, 2, 3 [ex. '1, 2, 3']: ").split(", ")

# Can also be put in the print function;
# this one is to prittify and escape 80 characters limit.
result = int(intSets[0]) + int(intSets[1]) + int(intSets[2])

print(f"{intSets[0]} + {intSets[1]} + {intSets[2]} = {result}")
````

Q7. Write a program that asks the user to enter a password. Then, check if the password is at least 8 characters long and the password contains only uppercase letters

Q8. Write a program that asks the user to enter the radius of a circle. Then, calculate and display the area and circumference of the circle. (Use the math module)

Formula:

- Circumference = 2 × π × r
- Area = π × r²

````python
import math # Can also use `import math as m` to use math module as `m`

radius = int(input("Radius: "))

print(f"Circumference: {2 * math.pi * radius} / Area: {math.pi * math.pow(radius, 2)}")
````

Q9. Write a program that asks the user to enter a temperature in Celsius. Then:

- Convert the temperature to Fahrenheit
- Convert the temperature to Kelvin
- Display both converted values

````python
cel = int(input("Temp in Celcius: "))

fah = (cel * (9/5)) + 32   # Formula: (C * (9/5)) + 32
kel = cel + 273.15         # Formula: C + 273.15

print(f"Fahrenheit: {fah} / Kelvin: {kel}")
````

Q10. Write a program that receives a duration in seconds from the user and convert it in the format shown below:
```
<hour>:<min>:<sec>
```

Ex.
```python
Enter duration: 3690<enter>
1:1:30
```

````python
data = int(input("Enter time in seconds [ex. 3690]: "))

sec = data % 60
min = round(((data - sec) / 60) % 60)
hrs = round((data - (min * 60) - sec) / 3600)

print(f"{hrs}:{min}:{sec}")
````
