Q5. Write a program that asks the user to enter a positive number. Then, calculate and display the factorial of that number.

Ex:
5! = 5×4×3×2×1 = 120

````python
number = int(input("Number to factorial (n!): "))

string = ""
sum = 0

# Catch invalid cases.
if (number < 0):
  print("This algorithm only accepts positive integers.")
else:
  if (number == 0):
    string = "0×"
  else:
    sum = 1 # Integers can't be multiplied by 0 to get a value.
    for i in range(number):
      sum = sum * (number - i)
      string = string + f"{number - i}×"

  print(f"{string[:-1]} = {sum}")
````

Q7. Write a program that keeps asking the user to enter a number until they enter -1. Then, display the total sum of all the numbers entered (excluding the -1).

````python
# Code here

keepRunning = True
sum = 0

while (keepRunning == True):
  data = int(input("Keep entering number (End the program with '-1'): "))
  if (data == -1):
    keepRunning = False
    print(sum)
    continue
  sum = sum + data
````

Q9. Write a program that asks the user to enter a number n. Then, display the pattern below using a for loop:

Ex (if n = 5):
```
*
**
***
****
*****
*****
****
***
**
*
  *
 ***
*****
```

````python
import math

number = int(input("Enter pattern length: "))
tree = math.floor(number / 2)

for i in range(number):
  print("*" * (i + 1))
for i in range(number):
  print("*" * (number - (i)))

while tree >= 0:
  if (number - tree * 2 != 0):
    print((" " * tree) + ("*" * (number - tree * 2)))
  tree = tree - 1
````

Q10. Write a program that recieves an integer, n, from the user, which is the Fibonacci number, then print out the last value of the Fibonacci sequence where each number in the sequence is reated as follows.
* Fibonacci(0) = 1
* Fibonacci(1) = 1
* Fibonacci(n) = Fibonacci(n-1) + Fibonacci(n-2)

Ex.
```
Enter n : 0<enter>
Fibonacci(0) = 0
```
```
Enter n : 10<enter>
Fibonacci( 10 ) = 55
```

````python
number = int(input("Enter Fibonnacci number: "))

a, b = 0, 1
for i in range(number):
  a, b = b, a + b

print(f"Fibonacchi({number}) = {a}")
````
