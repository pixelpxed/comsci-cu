1. Write a program that computes the net amount of a balance based on a transaction log from console input. The transaction log is contained in one line in the format of [W|D number ]* e.g. D 300 D 300 W 200 D 300. W means withdrawal and D means deposit. The output for the above example input would be 700.


```python
def get_balance(instructions):
  balance = 0
  instructions = instructions.split(" ")

  # Assume that file has no mistakes, else there would be odd number of arrays.
  for i in range(int(len(instructions) / 2)):
    operation = instructions[i * 2]
    amount = int(instructions[(i * 2) + 1])
    
    if   (operation == "D"): balance = balance + amount
    elif (operation == "W"): balance = balance - amount

  print(balance)

# Tests
get_balance("D 300 D 300 W 200 D 300 W 200") # 500
get_balance("D 500 W 100 D 200 W 400 D 250") # 450
```

2. A robot moves in a plane starting from the origin (0,0). The robot can move toward UP (U), DOWN (D), LEFT (L) and RIGHT(R)with a given input e.g. U5D3L3R2. The number after the direction are steps. Compute the distance from the current position of a robot after a sequence of moves.

```python
import math

# Fix: Accepts double digit values.
def get_distance(instructions):
  x, y = [0, 0]
  # Assume that file has no mistakes, else there would be odd number of arrays.
  for i in range(int(len(instructions) / 2)):
    operation = instructions[i * 2]
    amount = int(instructions[(i * 2) + 1])

    if   (operation == "U"): y = y + amount
    elif (operation == "D"): y = y - amount
    elif (operation == "R"): x = x + amount
    elif (operation == "L"): x = x - amount

  # Formula: sqrt((x1 - x2)^2 + (y1 - y2)^2)
  # Assume starting position is 0, we cut x2, y2 out.
  return print(math.sqrt((x ** 2) + (y ** 2)))

# Tests
get_distance("U5D3L3R2")   # 2.236...
get_distance("U9D4R2R2L1") # 5.830...
```

3. Write a program that check the validity of a given password. A password is valid if it contains at least 1 occurence of the character c and exactly 1 occurrence of the character 1. For example, a password ABcd0123bca$ is valid, while a password c11 is invalid.

```python
def validate_password(password):
  count_c, count_1 = [0, 0]

  password = password.lower()

  for i in range(len(password)):
    if   (password[i] == "c"): count_c = count_c + 1
    elif (password[i] == "1"): count_1 = count_1 + 1

  return (count_c >= 1 and count_1 == 1)

# Tests
print(validate_password("ABcd0123bca")) # True
print(validate_password("c11"))         # False
```

4. **Protein Length Summary**:

5. **DNA Sequence Analyzer:** Write a program that helps a lab technician analyze a DNA sequence.
The user will enter a DNA string containing the letters A, T, G, and C.
Your program should report:
- the total number of nucleotides,
- how many times each base (A, T, G, C) appears, and
- the GC content percentage (the percentage of nucleotides that are either G or C).

```python
import math

def analyze_sequence(sequence):
  a, t, c, g = [0, 0, 0, 0]

  for i in range(len(sequence)):
    if (sequence[i] == "A"):
      a = a + 1
    elif (sequence[i] == "T"):
      t = t + 1
    elif (sequence[i] == "C"):
      c = c + 1
    elif (sequence[i] == "G"):
      g = g + 1

  print(f"""{sequence}
 - Total Nucleotides: {a + t + c + g}")
 - ATCG Appearances:
    - A: {a} times
    - T: {t} times
    - C: {c} times
    - G: {g} times
 - GC Percentage: {math.ceil((c + g) / ((a + t + c + g) / 100) * 100) / 100}%
""")

analyze_sequence("ATGC")      #  50.00%
analyze_sequence("AAAA")      #  00.00%
analyze_sequence("GGCC")      # 100.00%
analyze_sequence("ATGCCGTA")  #  50.00%
analyze_sequence("TTAACCGGG") #  55.56%
```

6. Accept top left corner coordinate (x,y), width and height of two rectangles e.g. [5,5,10,10] and [0,0,12,15]. Then print “yes” if they intersect otherwise print ”no”

```python
def check_intersect(r1, r2):
  # List all positions in the direction of each square.
  r1_x_list = list(range(r1[0], r1[0] + r1[2] + 1))
  r1_y_list = list(range(r1[1], r1[1] + r1[3] + 1))
  r2_x_list = list(range(r2[0], r2[0] + r2[2] + 1))
  r2_y_list = list(range(r2[1], r2[1] + r2[3] + 1))

  # Check for intersecting positions in each direction.
  x_hit_len = len(list(set(r1_x_list) & set(r2_x_list)))
  y_hit_len = len(list(set(r1_y_list) & set(r2_y_list)))

  # Rectangles intersect if both directions hits something.
  return print("yes" if (x_hit_len != 0 and y_hit_len != 0) else "no")

# Tests
check_intersect([5, 5, 10, 10], [0, 0, 12, 15]) # yes
check_intersect([5, 5, 10, 10], [0, 0, 2, 4])   # no
```

7. Accept two integer lists as input and check if they are circularly identical. Print ”yes” if they are, print ”no” if they are not.


```python
def circular_equal(int1, int2):
  int1 = str(int1)
  int2 = str(int2)

  # Make a circle string.
  int2 = int2 + int2

  return print("yes" if (int1 in int2) else "no")

# Tests
circular_equal(123, 312)   # yes
circular_equal(1234, 3412) # yes
circular_equal(1234, 1234) # yes
circular_equal(123, 321)   # no
circular_equal(1234, 1233) # no
```
