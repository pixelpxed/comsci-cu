Q1. Write a program to read a file called numbers.txt, where each line contains a number. Display the total sum of all the numbers.

````python
# Expect: 45

filePath = open("numbers.txt")
contents = filePath.read().split("\n")

sum = 0
for i in range(len(contents)):
  sum = sum + int(contents[i])
print(sum)

filePath.close()
````

Q2. Write a program to read a file called records.txt, where each line contains a name followed by a score (e.g. Alice 85). Display only the names of students who scored 50 or higher.

Q3. Write the function named `sum_digits` that accepts an integer and returns the summation of each digit.

````python
def sum_digits(integer):
  integer = str(integer)

  sum = 0
  for i in range(len(integer)):
    sum = sum + int(integer[i])

  return sum

sum_digits(1234) # TEST, expect 10
````

Q5. Write the function named `check_date` that accepts three integer numbers `(m,d,y)` where `y` is the year in AD and returns if the given date is valid. Note that leap years are the years that are multiples of four but not are multiples of 100 or the years that are multiples of 400.

````python
# Assuming February is not a leap year.
days_in_month = [
  31, 28, 31, 30, 31, 30,
  31, 31, 30, 31, 30, 31
]

def check_date(m, d, y):
  # Not accounting for February leap year!
  data = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

  # Get days in month from the `data` array.
  expected_days = data[m - 1]

  # Accounting for leap year edge cases.
  if (m == 2) and (  # If February.
    ((y % 4 == 0) and (y % 100 != 0)) or 
    (y % 400 == 0)
  ):
    expected_days = 29

  if (d <= expected_days):
    return True
  else:
    return False

check_date(2, 29, 2000) # TEST: 29/02/2000, expected: True
````

Q6. Write a function that accepts 3 real numbers and returns the lowest value.


````python
def lowest_val(a, b, c):
  array = [a, b, c]

  lowest = array[0]
  for i in range(len(array) - 1):
    if array[i] > array[i + 1]:
      lowest = array[i + 1]

  return lowest

lowest_val(5, 4, 6) # TEST, expect 4
````

Q7. Write a function that accepts a number `n` and returns the value of `n!`.


````python
def factorial(n):
  if (n == 0):
    return 0
  if (n < 0):
    return "Invalid input!"

  sum = 1
  for i in range(n):
    sum = sum * (n - i)

  return sum

factorial(5) # TEST, expect 120
````

Q8. Write a function that accepts 4 realed numbers `x1,x2,x3,x4` which represent the coordinates (x1,y1) and (x2,y2) to calculate and return the distance between these two points.


````python
import math

def point_distance(x1, y1, x2, y2):
  return (
    math.sqrt(((x1 - x2) ** 2) + ((y1 - y2) ** 2))
  )

point_distance(2, 3, 5, 7) # TEST: (2, 3) and (5, 7), expect 5.0
````

````python
import math

def point_distance(x1, y1, x2, y2):
  return (
    math.sqrt(((x1 - x2) ** 2) + ((y1 - y2) ** 2))
  )

def area_by_point_distance(x1, y1, x2, y2):
  return (
    math.pi * (point_distance(x1, y1, x2, y2) ** 2)
  )

area_by_point_distance(2, 3, 5, 7) # TEST: (2, 3) and (5, 7), expect 78.539...
````

Q10. Write a function that accepts m and n from the user to calculate $C(m,n) = \dfrac{m!}{n!(m-n)!}$ by using the factorial function already written above.

````python
def factorial(n):
  if (n == 0):
    return 0
  if (n < 0):
    return "Invalid input!"

  sum = 1
  for i in range(n):
    sum = sum * (n - i)

  return sum

def combinations(m, n):
  return (
    factorial(m) / (factorial(n) * factorial(m - n))
  )

combinations(4, 2) # TEST: C(4, 2), expect: 6.0
````
