Python จะชี้ว่าโค้ดผิดตรงไหน พร้อมบอกประเภท:

`SyntaxError: ...`
- เขียนอะไรไม่ครบ เช่น ลืมปิด `)`, ลืมใส่ `:` หลัง if else elif while, ฯลฯ

`TypeError: ... 'ชื่อประเภท' cannot ...`
- ประเภทตัวแปรใช้ไม่ได้ ลองเปลี่ยนประเภทดู
- เช่น `float` เอาไปใช้ใน `range(1.0)` ไม่ได้ ต้องเป็น `int` `range(1)`

`NameError: name "variable" is not defined`
- ไม่เคยประกาศชื่อตัวนี้มาก่อน โปรแกรมไม่รู้คืออะไร
- ไปประกาศชื่อตัวแปรมาก่อน `variable = "สักอย่าง"` ฯลฯ

`AssertionError:`
- ซ้าย ขวา ไม่เท่ากัน เจออันนี้บ่อยตอนเทส
- เช่น `assert is_prime(1) == False` รัน `is_prime(1)` แล้วได้ `True`
- ลองรันแค่ Function ดู แล้วดูค่าที่ได้ ต่างกับผลลัพธ์ยังไง เช่น
  - `print(is_prime(1))` แล้วลองดูว่าผลลัพธ์ผิดตรงไหน แล้วไล่โค้ดดู

`ZeroDivisionError: division by zero`
- หารด้วย 0 (ทำไม่ได้)

`ValueError: math domain error`
- ตัวเลขมันเอาไปทำต่อไม่ได้ `undefined`
- พบบ่อยกับ `math.sqrt()` แล้วค่าข้างในเป็นติดลบ (ทำไม่ได้)

1. A bodybuilder keeps a log of the maximum reps he can push each day. He wants to know his personal best rep and his longest increasing streak.

Input:
- Take multiple lines of input up until `end` is inputted.
- Each line of input is an integer, represeting a maximum rep in the bodybuilder's log.

Output:
- If the log of reps is empty, output `No Data`.
- Else, the first line is the longest increasing streak (i.e., 100, 110, 120, but NOT 100, 110, 110).
- The second line is the bodybuilder's personal best of reps, irrespective to the streak.

Examples:
- Example 1:
```
100
120
130
end
```
```
3
130
```
- Example 2:
```
10
20
30
20
30
40
end
```
```
3
40
```
- Example 3:
```
100
100
90
80
70
end
```
```
1
100
```
- Example 4:
```
end
```
```
No Data
```

````python
reps_data, reps_input = [[], ""]
streaks, streak_count = [[], 0]

while (reps_input != "end"):
  reps_input = input("Enter Reps: ")
  if (reps_input != "end"):
    reps_data.append(int(reps_input))

if (len(reps_data) == 0):
  print("No Data")
else:
  for i in range(1, len(reps_data)):
    if (reps_data[i - 1] < reps_data[i]):
      streak_count = streak_count + 1 # Can also write as `streak_count += 1`
    else:
      streaks.append(streak_count)
      streak_count = 1
  streaks.append(streak_count) # Inserts the final streak count.

  print(max(streaks))
  print(max(reps_data))
````

## Examples

````python
denominators = [(100, 0), (50, 0), (20, 0), (10, 0), (5, 0), (1, 0)]
number = int(input("Enter number to denominate: "))
for i in range(len(denominators)):
  denominators[i] = (denominators[i][0], number // denominators[i][0])
  number = number - (denominators[i][0] * denominators[i][1])
print(" ".join(list(str(j[1]) for j in denominators)))
````

2. Set the even position of a 5, length array to 0.

````python
import math

array = [1, 2, 3, 4, 5]

def setZeroEvenPos(data):
  for i in range(math.ceil(len(data) / 2)):
    data[i * 2] = 0
  return data

print("Set Zero Even Positions:", setZeroEvenPos(array))
````

1.
Write a function that accepts a list of numbers and returns the average; implement it in two ways: using `for ... in list` and `for ... in range`.

````python
array = [1, 2, 3, 4, 5, 6, 7, 8] # Expect: 4.5 (Avg.)

def listMethod(data):
  sum = 0
  for num in data:
    sum = sum + num
  return sum / len(data)

def rangeMethod(data):
  sum = 0
  for i in range(len(data)):
    sum = sum + data[i]
  return sum / len(data)

print("List Method:", listMethod(array))
print("Range Method:", rangeMethod(array))
````

2. Write a function that accepts a list of integers, then returns a sum of values except the values in between "6" to "7". (Defines that for every number "6", it must be followed by "7" somewhere in the array later on.)

Examples:
  - `sum67([1, 2, 2])` = 5
  - `sum67([1, 2, 2, 6, 99, 99, 7])` = 5
  - `sum67([1, 1, 6, 7, 2])` = 4
  - `sum67([1, 6, 2, 2, 7, 1, 6, 99, 99, 7])` = 2
  - `sum67([1, 6, 2, 6, 2, 7, 1, 6, 99, 99, 7])` = 2
  - `sum67([2, 7, 6, 2, 6, 7, 2, 7])` = 18
  - `sum67([2, 7, 6, 2, 6, 2, 7])` = 9
  - `sum67([1, 6, 7, 7])` = 8
  - `sum67([6, 7, 1, 6, 7, 7])` = 8
  - `sum67([6, 8, 1, 6, 7])` = 0
  - `sum67([])` = 0
  - `sum67([6, 7, 11])` = 11
  - `sum67([11, 6, 7, 11])` = 22
  - `sum67([2, 2, 6, 7, 7])` = 11

````python
def sum67(data):
  sum, add = 0, True
  for i in range(len(data)):
    if (data[i] == 6): add = False
    if (add == True):  sum = sum + data[i]
    if (data[i] == 7): add = True
  return sum

# Tests
assert sum67([1, 2, 2]) == 5
assert sum67([1, 2, 2, 6, 99, 99, 7]) == 5
assert sum67([1, 1, 6, 7, 2]) == 4
assert sum67([1, 6, 2, 2, 7, 1, 6, 99, 99, 7]) == 2
assert sum67([1, 6, 2, 6, 2, 7, 1, 6, 99, 99, 7]) == 2
assert sum67([2, 7, 6, 2, 6, 7, 2, 7]) == 18
assert sum67([2, 7, 6, 2, 6, 2, 7]) == 9
assert sum67([1, 6, 7, 7]) == 8
assert sum67([6, 7, 1, 6, 7, 7]) == 8
assert sum67([6, 8, 1, 6, 7]) == 0
assert sum67([]) == 0
assert sum67([6, 7, 11]) == 11
assert sum67([11, 6, 7, 11]) == 22
assert sum67([2, 2, 6, 7, 7]) == 11
print("All Tests Passed!")
````

````python
def add_item(item_list, item):
    if (item in item_list):
        return "Item is already in the list"
    item_list.append(item)
    return "Item has been added"

def change_item(item_list, old_item, new_item):
    for i in range(len(item_list)):
        if (item_list[i] == old_item):
            item_list[i] = new_item
            return "Item has been changed"
    return "Item is not in the list"

def insert_item(item_list, item, location):
    item_list.insert(location, item)
    return "Item has been inserted"

def remove_item(item_list, item):
    if (item in item_list):
        item_list.remove(item)
        return "Item has been removed"
    return "This item is not in the list"

def show_items(item_list):
    if (len(item_list) != 0):
        return item_list
    return "The list is currently empty"

item_list = []

def Main():
    print("=== Shopping List Menu ===")
    print("1. Show items")
    print("2. Add item")
    print("3. Change item")
    print("4. Insert item")
    print("5. Remove item")
    print("6. Exit")
    while True:
        menu = int(input("Select menu: "))

        if menu == 1:
            print(show_items(item_list), end="\n\n")
        elif menu == 2:
            item = input("Enter item: ")
            print(add_item(item_list, item), end="\n\n")
        elif menu == 3:
            old_item = input("Enter old item: ") 
            new_item = input("Enter new item: ")
            print(change_item(item_list, old_item, new_item), end="\n\n")
        elif menu == 4:
            item = input("Enter item: ")
            location = int(input("Enter location index: "))
            print(insert_item(item_list, item, location), end="\n\n")
        elif menu == 5:
            item = input("Enter item: ")
            print(remove_item(item_list, item), end="\n\n")
        elif menu == 6:
            return print("Goodbye!")
````
